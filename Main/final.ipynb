<h2 style="color:#FFA500";> 3. Target Variable — Cleaning & Binary Label </h2>

Clean `O_TC` (total tooth count) and produce a binary classification target:

- **a.** Most missing `O_TC` entries appear as `NaN`, but the 2024 file (HN24) encodes them as a `-1.80E+308` overflow sentinel that parses to `-inf` rather than `NaN`.
- **b.** `pd.to_numeric(..., errors='coerce')` coerces any remaining non-numeric values to `NaN`; the `±inf` sentinels are then replaced explicitly with `NaN`.
- **c.** Rows where `O_TC` is `NaN` are dropped; only records with a valid tooth count are kept.
- **d.** Binary label `O_TC_`: `1` if tooth count < 20 (functionally impaired), `0` otherwise.
- **e.** Class imbalance is assessed via the ratio of majority (0) to minority (1) class counts.


```python
print(df_subset[pd.to_numeric(df_subset['O_TC'], errors='coerce').isna()]['O_TC'].unique())
# ^ finds all non numeric codes in the O_TC column and displays what those 'errors' were, in this case it 
# was ['-1.80E+308']

df_subset['O_TC'] = pd.to_numeric(df_subset['O_TC'], errors='coerce')
# -1.80E+308 overflows float64 and becomes -inf (not NaN), so replace both infinities explicitly
df_subset['O_TC'] = df_subset['O_TC'].replace([np.inf, -np.inf], np.nan)

before_drop = len(df_subset)
df_subset = df_subset.dropna(subset=['O_TC'])
print(f"Rows dropped (NaN O_TC): {before_drop - len(df_subset)}")
# ^ to know how many rows were dropped because of NaN; not required to the code --  just to visualize

df_subset['O_TC_'] = (df_subset['O_TC'] < 20).astype(int)
# ^ a lot of research papers determined O_TC value below 20 is considered impactful to the normal
# chewing and digestion problems. 
# "astype(int) --> converts True = 1 and False = 0"

label_counts    = df_subset['O_TC_'].value_counts()
# ^ counts how many values are there in the new 'O_TC_' column of 0s and 1s 
label_0         = label_counts[0]
label_1         = label_counts[1]

print(f"Age filter: >= {age_threshold}")
print(f"Label 0 (≥20 teeth): {label_0}")
print(f"Label 1 (<20 teeth): {label_1}")
print(df_subset['O_TC'].value_counts(dropna=False).head(20))

print(f"NaN count: {df_subset['O_TC'].isna().sum()}")
```

<h2 style="color:#FFA500";> 4. Missing Value Audit & Invalid-Code Replacement </h2>

Survey responses use reserved codes (8, 9, 88, 99, 888, 999) for refusals and non-applicable items — these are **not** true values and must be replaced with `NaN` before modelling.

- **Disease columns** (`DE1_dg`): replace invalid codes (8 = N/A, 9 = missing) with `NaN`.
- **Survey columns** (`edu`, `ho_incm`, `occp`, `marri_1`, `sm_presnt`, `BP1`, `O_pain`, `MO4_00`, `BM1_0`–`BM1_8`, etc.): replace invalid codes with `NaN`.
- **Lab columns** (`HE_HbA1c`, `HE_BMI`, `HE_crea`, `HE_Uacid`, `HE_TG`, `HE_sbp`, `HE_dbp`, `HE_Ucrea`): coerce to numeric and nullify sentinel magnitudes (≥ 1e10).
- Creatinine values ≤ 0 are physiologically impossible and are also nullified.

```python
# Disease diagnoses: 8 = not applicable (children/adolescents), 9 = refused/missing
disease_cols = ['DE1_dg']

# Survey ordinal / binary: 8 = not applicable, 9 = refused/missing
# BD1 (lifetime drinking experience, >=1 full drink, excl. ceremonial sips):
#   1 = NEVER drank, 2 = Yes, 8 = N/A (child), 9 = unknown / no response.
#   NOTE THE DIRECTION -- 1 means never. Verified against the KNHANES skip pattern:
#   BD2_1 (drinking frequency) is 100% N/A when BD1 == 1 vs 29% when BD1 == 2.
#   Recoded to 1=drinker / 0=never in the modelling cells (map({1: 0, 2: 1})).
# HE_Upro (proteinuria): 9 = missing (valid values 0–5)
# O_pain: dental pain perception; 0=no pain, 1=pain
# MO4_00 (dental visit past year): 9 = missing
# BM1_0 (toothbrushing yesterday): 9 = missing
# BM1_1–BM1_8 (brushing timing): 8 = N/A (if BM1_0 == 0), 9 = missing
# BS12_1 (indoor secondhand smoke at workplace, past week): 1=yes, 2=no; 8=N/A, 9=missing
survey_cols = ['edu', 'ho_incm', 'occp', 'marri_1', 'house', 'sm_presnt', 'BD1', 'pa_aerobic', 'BP1', 'BS12_1', 'L_BR_FQ', 'HE_Upro', 'OR1', 'O_pain',
               'MO4_00', 'BM1_0', 'BM1_1', 'BM1_2', 'BM1_3', 'BM1_4', 'BM1_5', 'BM1_6', 'BM1_7', 'BM1_8']

# Lab continuous: coerce + overflow-sentinel removal (large float placeholders → NaN)
lab_cols = [
    'HE_HbA1c', 'HE_BMI', 'HE_crea',
    'HE_Uacid', 'HE_TG', 'HE_sbp', 'HE_dbp', 'HE_chol', 'HE_HDL_st2',
    'HE_Ucrea',
]

# Dietary continuous
diet_cols = ['N_CA', 'N_NA', 'N_PROT', 'N_TDF', 'LS_VEG1', 'LS_FRUIT', 'N_SUGAR']

# Coerce to numeric first: HN24 loads these columns as strings (its overflow-sentinel
# value forces object dtype), so the numeric-keyed .replace({8: nan, ...}) below would
# silently no-op on every 2024 row without this (mirrors the all_lab coercion pattern below).
df_subset[disease_cols + survey_cols] = df_subset[disease_cols + survey_cols].apply(pd.to_numeric, errors='coerce')

df_subset[disease_cols] = df_subset[disease_cols].replace({8: np.nan, 9: np.nan, 88: np.nan, 99: np.nan})
df_subset[survey_cols]  = df_subset[survey_cols].replace({8: np.nan, 9: np.nan, 88: np.nan, 99: np.nan})
df_subset[survey_cols]  = df_subset[survey_cols].replace([np.inf, -np.inf], np.nan)

all_lab = lab_cols + diet_cols
df_subset[all_lab] = df_subset[all_lab].apply(pd.to_numeric, errors='coerce')
df_subset[all_lab] = df_subset[all_lab].where(df_subset[all_lab].abs() < 1e10, np.nan)

df_subset.loc[df_subset['HE_crea'] <= 0, 'HE_crea'] = np.nan

# D13: nominal/count cols outside survey_cols still carried sentinels
# (verified in data: tins has 99, live_t has 9, all three have a stray -inf)
# Coerce to numeric first: HN24 loads these as strings too (same overflow-sentinel
# issue as disease_cols/survey_cols above), so cfam in particular was silently
# carrying the raw '-1.80E+308' string past the inf-replace below.
df_subset[['cfam', 'tins', 'live_t']] = df_subset[['cfam', 'tins', 'live_t']].apply(pd.to_numeric, errors='coerce')
df_subset[['cfam', 'tins', 'live_t']] = df_subset[['cfam', 'tins', 'live_t']].replace([np.inf, -np.inf], np.nan)
df_subset['tins']   = df_subset['tins'].replace({99: np.nan})   # 99 = missing (valid: 10/20/30)
df_subset['live_t'] = df_subset['live_t'].replace({9: np.nan})  # 9 = missing (valid: 1-4)
df_subset['cfam']   = df_subset['cfam'].replace({9: np.nan})    # 9 = missing sentinel (valid: 1-6)

# marri_2 / genertn / fam_rela: added nominal SES vars with COLUMN-SPECIFIC sentinels.
# They can't join the blanket survey_cols replace above because fam_rela's 9 is a VALID
# category (sibling), not a missing code -- a blanket {8,9,88,99}->NaN would destroy it.
# Coerce first (HN24 loads these as strings via the overflow sentinel, same as cfam/tins).
df_subset[['marri_2', 'genertn', 'fam_rela']] = df_subset[['marri_2', 'genertn', 'fam_rela']].apply(pd.to_numeric, errors='coerce')
df_subset[['marri_2', 'genertn', 'fam_rela']] = df_subset[['marri_2', 'genertn', 'fam_rela']].replace([np.inf, -np.inf], np.nan)
df_subset['marri_2']  = df_subset['marri_2'].replace({8: np.nan, 88: np.nan, 99: np.nan})  # 8/88 = N/A (unmarried), 99 = missing (valid: 1-4)
df_subset['genertn']  = df_subset['genertn'].replace({9: np.nan})                           # 9 = missing (valid: 1-7)
df_subset['fam_rela'] = df_subset['fam_rela'].replace({99: np.nan})                         # 99 = missing (valid: 1-10, incl. 9 = sibling)

# marri_2 gets EXPLICIT levels so marital status is fully encoded by ONE variable:
#   0   = never married (marri_1 == 2; marri_2 is structurally N/A for them)
#   1-4 = KNHANES detail (1 married-cohabiting, 2 married-separated, 3 widowed, 4 divorced)
#   5   = married but marital detail missing (genuine non-response among marri_1 == 1)
# marri_1 is then an exact function of marri_2 (marri_1 == 2 <=> marri_2 == 0), so it is
# NOT carried as a separate predictor in either arm -- the never-married contrast is read
# off the marri_2_0 dummy. Previously marri_1 + marri_2 together silently dropped all 380
# never-married respondents from every complete-case model AND left marri_1 zero-variance,
# where §11's inestimable-column guard removed it without comment.
df_subset.loc[df_subset['marri_1'] == 2, 'marri_2'] = 0
df_subset.loc[(df_subset['marri_1'] == 1) & df_subset['marri_2'].isna(), 'marri_2'] = 5
print("marri_2 levels after recode (0 = never married, 5 = married/detail missing):")
print(df_subset['marri_2'].value_counts(dropna=False).sort_index())
# region (1-17) and town_t (1/2): verified clean, no sentinel codes

print("NaN counts per column after cleaning:")
total = len(df_subset)
for col in disease_cols + survey_cols + lab_cols + diet_cols:
    count = df_subset[col].isna().sum()
    pct   = (count / total) * 100
    print(f"  {col:<15}: {count:>5}  ({pct:.1f}%)")
```

<h4 style="color:#7FDBFF";> 4.1.1 Preprocessing Screening — Missingness & Collinearity Across All Candidate Variables </h4>

Every variable ever considered for `raw_columns` -- the 42 currently kept, plus the previously hand-excluded ones whose stated reason was itself missingness or collinearity -- is screened here on this cohort's actual cleaned data, so each verdict is a computed number rather than an asserted one.

**Categories:**
- **Structural** (`year`, `O_TC`, `wt_itvex`, `psu`, `kstrata`, `age`, `sex`) are never candidates for exclusion and are skipped entirely.
- **Precursor** variables (`DE1_dg`, `HE_HbA1c`, `HE_crea`, `BM1_1`–`BM1_8`) are consumed by the eGFR / diabetes / brush_freq derivations below and then dropped or superseded -- auto-dropping them here would break those cells, so they are reported (missingness only) but exempt from the verdict.
- Everything else -- the **auto-decidable pool** -- gets a real verdict: **NaN% > 30 → exclude**; among the NaN survivors, **VIF > 10 → exclude**, **5 < VIF ≤ 10 → flag, keep**. Multi-category nominal variables (`region`, `occp`, `marri_2`, `genertn`, `fam_rela`, `tins`, `live_t`, `house`) skip the VIF pass -- their raw integer codes have no linear ordering, so VIF on them directly would be meaningless (their own collinearity is properly checked post-dummy-encoding in §11.3.2).

`OR1` (self-rated oral health) is now loaded and cleaned like any other variable here -- same treatment as `O_pain` and `MO4_00`: retained in the ML arm (prediction tolerates a downstream marker) and excluded from the §11 inferential exposures (etiological framing, reverse causation), with its own §9.5 sensitivity check.

**This cell only computes and prints verdicts -- it does not change `raw_columns`, `feature_cols` (cell 39), or `exposures` (cell 91).** Those keep their current hand-picked lists; review the table below and decide what to change there yourself.

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
import statsmodels.api as sm

# ── Category definitions ──────────────────────────────────────────────────────────────────
# Precursor: consumed by a derivation below, then dropped/superseded -- reported only.
precursor_cols = ['DE1_dg', 'HE_HbA1c', 'HE_crea',
                   'BM1_1', 'BM1_2', 'BM1_3', 'BM1_4', 'BM1_5', 'BM1_6', 'BM1_7', 'BM1_8']

# Already cleaned above (survey/lab/diet blocks) -- part of the auto-decidable pool. OR1 is
# here now too -- loaded and cleaned like O_pain/MO4_00 (see the survey_cols cell above).
already_clean_pool = [
    'ho_incm', 'edu', 'occp', 'marri_1', 'marri_2', 'genertn', 'fam_rela', 'house', 'cfam',
    'live_t', 'tins', 'region', 'town_t',
    'HE_BMI', 'HE_Uacid', 'HE_TG', 'HE_sbp', 'HE_dbp', 'HE_chol', 'HE_HDL_st2',
    'HE_Upro', 'HE_Ucrea',
    'sm_presnt', 'BD1', 'pa_aerobic', 'BP1', 'BS12_1',
    'N_CA', 'N_NA', 'N_PROT', 'N_TDF', 'LS_VEG1', 'LS_FRUIT', 'N_SUGAR', 'L_BR_FQ',
    'O_pain', 'MO4_00', 'BM1_0', 'OR1',
]
# Never loaded into df_subset -- pulled fresh from df below. Their ORIGINAL exclusion reason
# was itself missingness or collinearity, so their fate here is decided by the thresholds.
fresh_load_auto = {
    'HE_obe': 'lab', 'HE_wc': 'lab', 'HE_glu': 'lab', 'N_EN': 'lab', 'DI4_dg': 'lab',
    'DE1_4': 'disease', 'BS3_1': 'disease', 'BD2_1': 'disease', 'LW_ms': 'lab',
}
nominal_multi_category = {'region', 'occp', 'marri_2', 'genertn', 'fam_rela', 'tins', 'live_t', 'house'}

def _clean_disease(s):
    # Same {8,9,88,99}->NaN convention as disease_cols/survey_cols above.
    s = pd.to_numeric(s, errors='coerce')
    return s.replace({8: np.nan, 9: np.nan, 88: np.nan, 99: np.nan})

def _clean_lab(s):
    # Same coerce + overflow-sentinel convention as lab_cols/diet_cols above.
    s = pd.to_numeric(s, errors='coerce')
    return s.where(s.abs() < 1e10, np.nan)

# LW_ms gets ONLY the lab-style (coerce + overflow) cleaning, not the blanket {8,9,88,99}
# replace: its value 9 may be a genuine category rather than a missing code, the same
# fam_rela pitfall noted in §4 above. This is already >50% NaN/-inf structurally (N/A for
# males) regardless, so the choice does not change its verdict.

screen = df_subset[precursor_cols + already_clean_pool].copy()
for col, kind in fresh_load_auto.items():
    raw = df.loc[df_subset.index, col]
    screen[col] = _clean_lab(raw) if kind == 'lab' else _clean_disease(raw)

n_total = len(screen)
nan_pct = (screen.isna().sum() / n_total * 100).round(2)

# STAGE 1 (NaN, marginal): decides who is even eligible for the VIF stage. Running VIF
# jointly across variables with very different missingness patterns (e.g. BD2_1's "N/A for
# non-drinkers" alongside LW_ms's "N/A for males") collapses the complete-case intersection
# to almost nothing and makes every VIF explode to inf/undefined -- so anything that already
# fails the NaN test is dropped BEFORE the VIF matrix is built, not lumped in with it.
auto_pool = already_clean_pool + list(fresh_load_auto)
nan_survivors = [c for c in auto_pool if nan_pct[c] <= 30]

# STAGE 2 (VIF, joint): complete-case among the NaN-stage survivors only (nominal skipped).
vif_pool = [c for c in nan_survivors if c not in nominal_multi_category]
vif_frame = screen[vif_pool].dropna()
_zero_var = [c for c in vif_pool if vif_frame[c].nunique() <= 1]
_vif_cols = [c for c in vif_pool if c not in _zero_var]
_Xc = sm.add_constant(vif_frame[_vif_cols].astype(float), has_constant='add')
vif = pd.Series(
    [variance_inflation_factor(_Xc.values, i) for i in range(_Xc.shape[1])],
    index=_Xc.columns
).drop('const')

rows = []
for col in nan_pct.index:
    cat = 'precursor (exempt)' if col in precursor_cols else 'auto-decidable'
    row = {'Variable': col, 'Category': cat, 'NaN %': nan_pct[col]}
    if col in vif.index:
        row['VIF'] = round(vif[col], 2)
    elif col in nominal_multi_category and col in nan_survivors:
        row['VIF'] = 'skipped (nominal)'
    elif col in _zero_var:
        row['VIF'] = 'undefined (zero variance)'
    elif col in auto_pool and col not in nan_survivors:
        row['VIF'] = 'not computed (failed NaN stage)'
    else:
        row['VIF'] = np.nan
    rows.append(row)
verdict_df = pd.DataFrame(rows).set_index('Variable')

def _verdict(r):
    if r['Category'] != 'auto-decidable':
        return 'kept as-is (see cell above)'
    if r['NaN %'] > 30:
        return 'EXCLUDE (NaN% > 30)'
    v = r['VIF']
    if isinstance(v, (int, float)) and not pd.isna(v):
        if v > 10:
            return 'EXCLUDE (VIF > 10)'
        if v > 5:
            return 'FLAG (5 < VIF <= 10), keep'
    return 'keep'

verdict_df['Verdict'] = verdict_df.apply(_verdict, axis=1)
print(f"Preprocessing screening -- n = {n_total} cohort rows, "
      f"VIF complete-case n = {len(vif_frame)} among {len(_vif_cols)} numeric survivors")
print(verdict_df.sort_values(['Category', 'NaN %']).to_string())

_excluded_vif = verdict_df[(verdict_df['Category'] == 'auto-decidable') & (verdict_df['Verdict'] == 'EXCLUDE (VIF > 10)')]
if not _excluded_vif.empty:
    print(f"\nNote: {list(_excluded_vif.index)} crossed VIF > 10 because they were tested "
          f"jointly with their own near-redundant counterpart already in the pool (e.g. "
          f"HE_BMI vs. HE_obe, which was excluded specifically for being redundant with "
          f"HE_BMI). A high VIF here says 'one of this pair should go', not necessarily "
          f"which one -- read this table, don't apply it mechanically.")
```

<h3 style="color:yellow";> 4.2 Derived Feature — eGFR (CKD-EPI 2021) </h3>

Replace raw serum creatinine with estimated Glomerular Filtration Rate (eGFR) using the **CKD-EPI 2021 creatinine equation**. eGFR is a more clinically interpretable measure of kidney function than creatinine alone and reduces the sex-dependence of the raw value.

`HE_crea` is dropped after the derivation; `sex` is retained as a model feature and inferential covariate.

2021 CKD-EPI Creatinine Equation
The 2021 CKD-EPI eGFRcr equation estimates GFR using age, sex, and serum creatinine.2

The equation can be expressed as a single equation:

<h4 style="color:#7FDBFF";> eGFR = 142 × min(SCr/κ, 1)^α × max(SCr/κ, 1)^(-1.200) × 0.9938^Age × 1.012 (if female) </h4>

Abbreviations/units

-   eGFR = estimated GFR in mL/min/1.73 m2
-   SCr = standardized serum creatinine in mg/dL
-   κ = 0.7 (females) or 0.9 (males)
-   α = -0.241 (females) or -0.302 (males)
-   min = indicates the minimum of SCr/κ or 1
-   max = indicates the maximum of SCr/κ or 1
-   age = years

source: https://www.niddk.nih.gov/research-funding/research-programs/kidney-clinical-research-epidemiology/laboratory/glomerular-filtration-rate-equations/adults

```python
def calculate_egfr(row):
    scr = row['HE_crea']
    age = row['age']
    sex = row['sex']   # KNHANES: 1 = male, 2 = female

    if pd.isna(scr) or pd.isna(sex):
        return np.nan

    if sex == 2:   # female
        kappa, alpha, sex_factor = 0.7, -0.241, 1.012
    else:          # male
        kappa, alpha, sex_factor = 0.9, -0.302, 1.0

    ratio = scr / kappa
    return (142
            * (min(ratio, 1) ** alpha)
            * (max(ratio, 1) ** -1.200)
            * (0.9938 ** age)
            * sex_factor)

df_subset['eGFR'] = df_subset.apply(calculate_egfr, axis=1)
df_subset = df_subset.drop(columns=['HE_crea'], errors='ignore')
# sex is retained — used as a feature in the ML model and as a covariate in the inferential model
```

<h3 style="color:yellow;"> 4.3 Derived Behavioral & Clinical Flags — Diabetes, Hypertension & Brushing </h3>

Diabetes, hypertension, and brushing frequency are derived into analysis-ready flags below, with summary counts printed instead of separate distribution plots.

<h4 style="color:lightblue;"> Summary — Diabetes, Hypertension & Brushing Frequency </h4>

**Diabetes & hypertension.** Both are derived binary flags (1 = condition present, 0 = absent). NaN rows (both source fields missing) are excluded from the printed counts. Percentages are of the valid (non-NaN) population.

**Brushing frequency.** `brush_freq`: sum of the BM1_1–BM1_8 occasion flags (0–8); `0` = did not brush yesterday (`BM1_0 == 0`), `NaN` = predominantly `BM1_0` itself missing (median-imputed downstream); a small residual reflects rows where `BM1_0` is answered but the occasion flags are still unrecorded. Both `BM1_0` and `brush_freq` are used exactly as KNHANES records them, with no separate handling for edentulous respondents and none dropped.

**Edentulism overlap.** Respondents with no natural teeth (`O_TC == 0`) are tooth-loss cases by construction (`O_TC == 0` ⇒ `O_TC_ == 1`), so any overlap with the brushing variables puts a structural component into the brushing estimate. That overlap is real but limited, and it is *not* concentrated among non-brushers — the majority of edentulous respondents report brushing anyway (dentures, gums, tongue). The cell above prints the exact counts per `BM1_0` level together with the target-positive rate recomputed **excluding** edentulous respondents, so the size of the structural component is visible rather than assumed. In this cohort the brushing gap persists clearly after their removal, so the association is not primarily an edentulism artefact — but note the non-brushing group is small, so the estimate for that level is imprecise.

```python
# Combined diabetes: diagnosed (DE1_dg == 1) OR HbA1c ≥ 6.5%
# NaN comparisons return False, so OR logic is safe — only set NaN when both sources are missing
diagnosed = (df_subset['DE1_dg'] == 1)
hba1c_pos = (df_subset['HE_HbA1c'] >= 6.5)
df_subset['diabetes'] = (diagnosed | hba1c_pos).astype(float)

both_missing = df_subset['DE1_dg'].isna() & df_subset['HE_HbA1c'].isna()
df_subset.loc[both_missing, 'diabetes'] = np.nan
df_subset = df_subset.drop(columns=['DE1_dg'], errors='ignore')

# Hypertension binary: SBP ≥ 140 OR DBP ≥ 90 (standard clinical threshold)
# NaN in either BP column → NaN result
sbp_pos = (df_subset['HE_sbp'] >= 140)
dbp_pos = (df_subset['HE_dbp'] >= 90)
df_subset['hypertension'] = (sbp_pos | dbp_pos).astype(float)

both_bp_missing = df_subset['HE_sbp'].isna() & df_subset['HE_dbp'].isna()
df_subset.loc[both_bp_missing, 'hypertension'] = np.nan

# BM1_1–BM1_8: binary flags for each brushing occasion → sum = brushing frequency count (0–8)
bm1_cols = ['BM1_1','BM1_2','BM1_3','BM1_4','BM1_5','BM1_6','BM1_7','BM1_8']
df_subset['brush_freq'] = df_subset[bm1_cols].sum(axis=1, min_count=1)
# Non-brushers (BM1_0 == 0) have all 8 slots coded N/A → NaN sum; their true frequency is 0.
df_subset.loc[df_subset['BM1_0'] == 0, 'brush_freq'] = 0
# Remaining brush_freq NaN = BM1_0 itself missing (median-imputed downstream)

# interdental dropped: BM2_1/BM2_3 are absent from the HN23 & HN24 files, so the feature
# is 100% missing for all of 2023 and the entire 2024 test year — degenerate, removed.

# Brushing is modelled exactly as KNHANES records it: BM1_0 (brushed yesterday, 0/1) and
# brush_freq (occasions yesterday, 0-8). Edentulous respondents are NOT given a separate
# category and are NOT dropped -- they answer BM1_0 like anyone else and stay in the sample.
#
# Edentulism does partly overlap with non-brushing (O_TC == 0 => O_TC_ == 1 by construction,
# so any edentulous respondent is a case), but it is NOT concentrated there: most edentulous
# respondents report brushing anyway -- plausibly dentures, gums or tongue. The table below
# prints the overlap and re-computes the brushing gap with edentulous respondents removed,
# so the structural component can be read off directly instead of assumed.
_edent   = (df_subset['O_TC'] == 0)
_nobrush = (df_subset['BM1_0'] == 0)
_brush   = (df_subset['BM1_0'] == 1)

# ── Diabetes & Hypertension summary ─────────────────────────────────────
print(f"{'diabetes = 1 (combined)':28s}{int(df_subset['diabetes'].sum()):>8,}")
print(f"{'diabetes = 0':28s}{int((df_subset['diabetes'] == 0).sum()):>8,}")
print(f"{'diabetes = NaN':28s}{df_subset['diabetes'].isna().sum():>8,}")
print(f"{'hypertension = 1':28s}{int(df_subset['hypertension'].sum()):>8,}")
print(f"{'hypertension = 0':28s}{int((df_subset['hypertension'] == 0).sum()):>8,}")
print(f"{'hypertension = NaN':28s}{df_subset['hypertension'].isna().sum():>8,}")

# ── Brushing frequency summary ───────────────────────────────────────
print(f"\n{'Edentulous respondents (O_TC == 0)':28s}{_edent.sum():>8,}"
      f"   of these, reporting brush_freq > 0: {(_edent & (df_subset['brush_freq'] > 0)).sum()}")
for _lbl, _m in [("BM1_0 == 0 (did not brush)", _nobrush), ("BM1_0 == 1 (brushed)", _brush)]:
    print(f"{_lbl:28s}{_m.sum():>8,}  edentulous within: {(_m & _edent).sum():5d}"
          f" ({(_m & _edent).sum() / max(_m.sum(), 1) * 100:5.1f}%)"
          f"  target-positive: {df_subset.loc[_m, 'O_TC_'].mean():6.1%}"
          f"  excl. edentulous: {df_subset.loc[_m & ~_edent, 'O_TC_'].mean():6.1%}")
print(f"{'brush_freq NaN':28s}{df_subset['brush_freq'].isna().sum():>8,}")
print(df_subset['brush_freq'].value_counts().sort_index())
```

```python
from scipy import stats

cont_vars = {
    "age"       : "Age (years)",
    "HE_HbA1c" : "HbA1c (%)",
    "eGFR"      : "eGFR (mL/min/1.73m²)",
    "HE_BMI"    : "BMI (kg/m²)",
    "HE_Uacid" : "Uric Acid (mg/dL)",
    "HE_sbp"    : "Systolic BP (mmHg)",
    "HE_dbp"    : "Diastolic BP (mmHg)",
    "HE_TG"     : "Triglycerides (mg/dL)",
    "HE_chol"   : "Total Cholesterol (mg/dL)",
    "HE_HDL_st2": "HDL Cholesterol (mg/dL)",
    "brush_freq": "Brushing Frequency",
}

# ── Table-1 scopes ───────────────────────────────────────────────────────────
# The three bivariate screens (§5.1-5.3) serve TWO distinct purposes, which the earlier
# train-only-only version conflated:
#
#   FULL (2020-2024)       -> the manuscript's Table 1. A Table 1 must describe the sample
#                             that was actually analysed, and §11's inferential arm fits all
#                             five waves with no hold-out, so omitting 2024 described a
#                             sample that no model in this notebook actually uses.
#   TRAIN-ONLY (2020-2023) -> the ML feature-selection screen. Variables enter feature_cols
#                             partly on the strength of these tests, so they must not be
#                             chosen using 2024 outcomes -- that would leak the hold-out into
#                             the ML arm. THIS is the real justification for restricting the
#                             years; the previous comment framed it as protecting §11, which
#                             it cannot, since §11 uses 2024 regardless.
#
# Cite the FULL table as Table 1; cite TRAIN-ONLY only when justifying ML variable inclusion.
# Defined locally in each screen cell so each runs standalone -- the same reason §9.3 and
# §10.2.1 re-derive their own sex vector.
TABLE1_SCOPES = {
    "FULL (2020-2024) — manuscript Table 1":        df_subset["year"].notna(),
    "TRAIN-ONLY (2020-2023) — ML selection screen": df_subset["year"] != 2024.0,
}

for _scope, _scope_mask in TABLE1_SCOPES.items():
    _s = df_subset[_scope_mask]
    g0 = _s[_s["O_TC_"] == 0]   # ≥20 teeth
    g1 = _s[_s["O_TC_"] == 1]   # <20 teeth

    rows = []
    for col, label in cont_vars.items():
        a = g0[col].dropna()
        b = g1[col].dropna()
        t, p = stats.ttest_ind(a, b, equal_var=False)
        sig = "***" if p < 0.001 else ("**" if p < 0.01 else ("*" if p < 0.05 else "ns"))
        rows.append({
            "Variable"       : label,
            "n (≥20 teeth)"  : len(a),
            "≥20 teeth"      : f"{a.mean():.2f} ± {a.std():.2f}",
            "n (<20 teeth)"  : len(b),
            "<20 teeth"      : f"{b.mean():.2f} ± {b.std():.2f}",
            "t"              : round(t, 3),
            "p-value"        : f"{p:.4f}",
            "Sig"            : sig,
        })

    ttest_df = pd.DataFrame(rows)
    print(f"\n══ Welch's t-test — {_scope}  (n = {len(g0) + len(g1):,}) ══")
    print(ttest_df.to_string(index=False))
```

<h3 style="color:yellow";> 5.2 Bivariate Comparison — Mann-Whitney U (Ordinal Variables) </h3>

Results reported as <b>Median [Q1–Q3]</b> per group.

```python
from scipy import stats

ord_vars = {
    "edu"     : "Education (1=≤Elem → 4=College)",
    "ho_incm" : "Income Quartile (1=Low → 4=High)",
    "HE_Upro" : "Proteinuria Severity (0–5)",
    "BP1"     : "Perceived Stress (1=High → 4=Low)",
    "L_BR_FQ" : "Breakfast Frequency (1=Daily → 4=Rarely)",
}

# ── Table-1 scopes ───────────────────────────────────────────────────────────
# The three bivariate screens (§5.1-5.3) serve TWO distinct purposes, which the earlier
# train-only-only version conflated:
#
#   FULL (2020-2024)       -> the manuscript's Table 1. A Table 1 must describe the sample
#                             that was actually analysed, and §11's inferential arm fits all
#                             five waves with no hold-out, so omitting 2024 described a
#                             sample that no model in this notebook actually uses.
#   TRAIN-ONLY (2020-2023) -> the ML feature-selection screen. Variables enter feature_cols
#                             partly on the strength of these tests, so they must not be
#                             chosen using 2024 outcomes -- that would leak the hold-out into
#                             the ML arm. THIS is the real justification for restricting the
#                             years; the previous comment framed it as protecting §11, which
#                             it cannot, since §11 uses 2024 regardless.
#
# Cite the FULL table as Table 1; cite TRAIN-ONLY only when justifying ML variable inclusion.
# Defined locally in each screen cell so each runs standalone -- the same reason §9.3 and
# §10.2.1 re-derive their own sex vector.
TABLE1_SCOPES = {
    "FULL (2020-2024) — manuscript Table 1":        df_subset["year"].notna(),
    "TRAIN-ONLY (2020-2023) — ML selection screen": df_subset["year"] != 2024.0,
}

for _scope, _scope_mask in TABLE1_SCOPES.items():
    _s = df_subset[_scope_mask]
    g0 = _s[_s["O_TC_"] == 0]
    g1 = _s[_s["O_TC_"] == 1]

    rows = []
    for col, label in ord_vars.items():
        a = g0[col].dropna()
        b = g1[col].dropna()
        u, p = stats.mannwhitneyu(a, b, alternative="two-sided")
        sig = "***" if p < 0.001 else ("**" if p < 0.01 else ("*" if p < 0.05 else "ns"))
        rows.append({
            "Variable"         : label,
            "n (≥20 teeth)"    : len(a),
            "Median [IQR] ≥20" : f"{a.median():.1f} [{a.quantile(0.25):.1f}–{a.quantile(0.75):.1f}]",
            "n (<20 teeth)"    : len(b),
            "Median [IQR] <20" : f"{b.median():.1f} [{b.quantile(0.25):.1f}–{b.quantile(0.75):.1f}]",
            "U"                : int(u),
            "p-value"          : f"{p:.4f}",
            "Sig"              : sig,
        })

    mwu_df = pd.DataFrame(rows)
    print(f"\n══ Mann-Whitney U — {_scope}  (n = {len(g0) + len(g1):,}) ══")
    print(mwu_df.to_string(index=False))
```

<h3 style="color:yellow";> 5.3 Bivariate Comparison — Chi-square (Categorical Variables) </h3>

Categorical variables (binary and nominal) are compared using Pearson chi-square.
Results reported as <b>n (%)</b> of the positive/reference category per group.
A † flag marks cells where any expected count &lt; 5 (Fisher’s exact preferred there).

<b>Binary variables</b> show % of the flagged category (Yes / positive code).<br>
<b>Nominal variables</b> (>2 unordered categories) show χ² and p-value only — too many categories for inline display.

**Reported for two scopes.** The **FULL (2020–2024)** table is the manuscript's Table 1 — it describes the sample actually analysed, since §11 fits all five waves. The **TRAIN-ONLY (2020–2023)** table is the ML feature-selection screen: variables entering `feature_cols` must not be chosen using 2024 outcomes, or the hold-out leaks into the ML arm. Cite the full table unless you are specifically justifying ML variable inclusion.

```python
from scipy import stats as sp_stats

def chi2_test(col, g0, g1):
    a = g0[col].dropna()
    b = g1[col].dropna()
    combined_group  = pd.concat([pd.Series([0]*len(a)), pd.Series([1]*len(b))], ignore_index=True)
    combined_values = pd.concat([a.reset_index(drop=True), b.reset_index(drop=True)], ignore_index=True)
    ct = pd.crosstab(combined_values, combined_group)
    chi2, p, dof, expected = sp_stats.chi2_contingency(ct)
    fisher_flag = "†" if (expected < 5).any() else ""
    sig = "***" if p < 0.001 else ("**" if p < 0.01 else ("*" if p < 0.05 else "ns"))
    return len(a), len(b), round(chi2, 3), dof, f"{p:.4f}", sig + fisher_flag

# ── Binary categorical ───────────────────────────────────────────────────────
# pos_key: the value to report as the "positive" category n(%)
binary_vars = {
    "sex"         : ("Sex (Female)",           2),
    "diabetes"    : ("Diabetes (Yes)",          1.0),
    "hypertension": ("Hypertension (Yes)",      1.0),
    "sm_presnt"   : ("Current Smoker (Yes)",    1),
    # BD1: 1 = NEVER drank, 2 = Yes (lifetime drinking experience). The positive
    # category is therefore 2, not 1 -- see the codebook note in §4.
    "BD1"         : ("Alcohol Drinker (Yes)",   2),
    "pa_aerobic"  : ("Aerobic PA (Yes)",        1),
    "BS12_1"      : ("Secondhand Smoke (Yes)",  1),
    "O_pain"      : ("Dental Pain (Yes)",       1),
    "MO4_00"      : ("Dental Visit past yr (Yes)", 1),
    "BM1_0"       : ("Brushed Yesterday (Yes)", 1),
    "town_t"      : ("Urban Area",              1),
}

# ── Nominal categorical ──────────────────────────────────────────────────────
nominal_vars = {
    "marri_1" : "Marital Status (ever/never married)",
    "marri_2" : "Marital Status (detailed, 0 = never married)",
    "occp"    : "Occupation",
    "cfam"    : "Household Composition",
    "live_t"  : "Living Arrangement",
    "house"   : "Housing Type",
    "tins"    : "Insurance Type",
    "region"  : "Region",
}

# ── Table-1 scopes ───────────────────────────────────────────────────────────
# The three bivariate screens (§5.1-5.3) serve TWO distinct purposes, which the earlier
# train-only-only version conflated:
#
#   FULL (2020-2024)       -> the manuscript's Table 1. A Table 1 must describe the sample
#                             that was actually analysed, and §11's inferential arm fits all
#                             five waves with no hold-out, so omitting 2024 described a
#                             sample that no model in this notebook actually uses.
#   TRAIN-ONLY (2020-2023) -> the ML feature-selection screen. Variables enter feature_cols
#                             partly on the strength of these tests, so they must not be
#                             chosen using 2024 outcomes -- that would leak the hold-out into
#                             the ML arm. THIS is the real justification for restricting the
#                             years; the previous comment framed it as protecting §11, which
#                             it cannot, since §11 uses 2024 regardless.
#
# Cite the FULL table as Table 1; cite TRAIN-ONLY only when justifying ML variable inclusion.
# Defined locally in each screen cell so each runs standalone -- the same reason §9.3 and
# §10.2.1 re-derive their own sex vector.
TABLE1_SCOPES = {
    "FULL (2020-2024) — manuscript Table 1":        df_subset["year"].notna(),
    "TRAIN-ONLY (2020-2023) — ML selection screen": df_subset["year"] != 2024.0,
}

for _scope, _scope_mask in TABLE1_SCOPES.items():
    _s = df_subset[_scope_mask]
    g0 = _s[_s["O_TC_"] == 0]
    g1 = _s[_s["O_TC_"] == 1]

    print(f"\n══ Chi-square — {_scope}  (n = {len(g0) + len(g1):,}) ══")

    rows_bin = []
    for col, (label, pos_key) in binary_vars.items():
        if col not in df_subset.columns:
            continue
        na, nb, chi2, dof, pval, sig = chi2_test(col, g0, g1)
        a = g0[col].dropna()
        b = g1[col].dropna()
        pct_a = f"{(a == pos_key).sum()} ({100*(a == pos_key).mean():.1f}%)"
        pct_b = f"{(b == pos_key).sum()} ({100*(b == pos_key).mean():.1f}%)"
        rows_bin.append({
            "Variable"    : label,
            "n (≥20)"     : na,
            "≥20 teeth"    : pct_a,
            "n (<20)"     : nb,
            "<20 teeth"   : pct_b,
            "χ²"          : chi2,
            "df"          : dof,
            "p-value"     : pval,
            "Sig"         : sig,
        })

    print("── Binary Categorical Variables (Chi-square) ──────────────────────────")
    print(pd.DataFrame(rows_bin).to_string(index=False))

    rows_nom = []
    for col, label in nominal_vars.items():
        if col not in df_subset.columns:
            continue
        na, nb, chi2, dof, pval, sig = chi2_test(col, g0, g1)
        rows_nom.append({
            "Variable" : label,
            "n (≥20)"  : na,
            "n (<20)"  : nb,
            "χ²"       : chi2,
            "df"       : dof,
            "p-value"  : pval,
            "Sig"      : sig,
        })

    print()
    print("── Nominal Categorical Variables (Chi-square) ──────────────────────────")
    print(pd.DataFrame(rows_nom).to_string(index=False))

print()
print("† Expected cell count < 5 — Fisher's exact preferred.")
```

<h2 style="color:#FFA500;"> 6. Train / Test Split (Temporal) & Mode/Median Imputation </h2>

Split by survey year to respect the temporal ordering of data collection:

| Set | Source files | Years |
|-----|-------------|-------|
| **Train** | HN20_all.csv, HN21_all.csv, HN22_all.csv, HN23_all.csv | 2020, 2021, 2022, 2023 |
| **Test** | HN24_all.csv | 2024 |

The imputer is fitted on the training fold only and applied to the test set without refitting, preventing data leakage. `year` is used solely as a split key and is excluded from `feature_cols`.

```python
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

# ML feature frame: work on a copy so df_subset stays intact for the inferential model
feat = df_subset.copy()

# B5: BD1 is LIFETIME DRINKING EXPERIENCE with 1 = NEVER and 2 = Yes (see §4 codebook
# note), so the drinker indicator is map({1: 0, 2: 1}) -- NOT {1: 1, 2: 0}. The earlier
# mapping inverted the variable, labelling never-drinkers as drinkers everywhere.
# O_pain and MO4_00 are already 0/1 in the data (verified in raw CSVs), left as-is.
feat['BD1'] = feat['BD1'].map({1: 0, 2: 1})
# B8: reverse BP1 so higher = more stressed (matches the inferential coding)
feat['BP1'] = 5 - feat['BP1']
# BS12_1: 1=exposed, 2=not exposed -> recode to 1/0 (binary)
feat['BS12_1'] = feat['BS12_1'].map({1: 1, 2: 0})

# marri_2 already carries its never-married (0) and married/detail-missing (5) levels,
# assigned once during cleaning in §4 -- no local recode here, and marri_1 is dropped
# from ordinal_cols below because it is an exact function of marri_2.

# B6/B7: one-hot encode nominal vars; trees unaffected, fixes LR
# (nominal codes have no interval order — raw scaled integers would mislead the LR model)
#
# REFERENCE LEVELS: one level per parent is dropped so the dummy block is not collinear
# with the intercept. Keeping every level (the previous dummy_na=False with no drop) made
# region -- 17 levels and ZERO missing values -- sum to exactly 1 in every row, an exact
# dummy trap. Trees were unaffected, but it left the Logistic Regression coefficients
# unidentified and therefore made its |coef| entries — used as an LR feature-importance signal in §9 — arbitrary. Reference levels match §11's inferential model so a coefficient
# means the same thing in both arms.
nominal_ref = {'tins': 10, 'region': 1, 'occp': 1, 'house': 1,
               'live_t': 1, 'marri_2': 1, 'genertn': 1, 'fam_rela': 1}
feat = pd.get_dummies(feat, columns=list(nominal_ref), prefix=list(nominal_ref),
                      dummy_na=True, dtype=float)
# dummy_na=True: a MISSING parent value gets its own explicit "unknown" column instead of
# an all-zero dummy row. Under the previous dummy_na=False that all-zero pattern was
# byte-identical to the dropped REFERENCE level, so respondents with an unknown occupation
# -- 8.8% of the >=50 cohort, 11.4% of >=65 -- were silently modelled as occp 1
# (Manager/professional & related). get_dummies also resolves NaN BEFORE the imputer runs,
# so the 'most_frequent' step never saw those cells and could not have fixed it.
# pandas emits a *_nan column for every parent even when it has no missing values, so drop
# the all-zero ones (region has zero missingness) to keep feature_cols honest.
_nan_cols  = [c for c in feat.columns if c.endswith('_nan')]
_empty_nan = [c for c in _nan_cols if feat[c].sum() == 0]
feat = feat.drop(columns=_empty_nan)
_unknown_kept = {c: int(feat[c].sum()) for c in _nan_cols if c not in _empty_nan}
_ref_cols = [f'{p}_{float(l)}' for p, l in nominal_ref.items()]
_missing_ref = [c for c in _ref_cols if c not in feat.columns]
assert not _missing_ref, f"Reference dummy column(s) not found: {_missing_ref}"
feat = feat.drop(columns=_ref_cols)

print("── Categorical Encoding " + "─" * 45)
print("Reference levels dropped (baseline):")
print("  " + ", ".join(_ref_cols))
print()
if _unknown_kept:
    print("Explicit 'unknown' categories kept (n rows):")
    for _col, _n in sorted(_unknown_kept.items(), key=lambda kv: -kv[1]):
        print(f"  {_col:<16s}{_n:>6,}")
    print()
if _empty_nan:
    print(f"All-zero 'unknown' columns dropped: {', '.join(_empty_nan)}")
    print()

tins_dummies     = sorted([c for c in feat.columns if c.startswith('tins_')])
region_dummies   = sorted([c for c in feat.columns if c.startswith('region_')])
occp_dummies     = sorted([c for c in feat.columns if c.startswith('occp_')])
house_dummies    = sorted([c for c in feat.columns if c.startswith('house_')])
live_t_dummies   = sorted([c for c in feat.columns if c.startswith('live_t_')])
marri_2_dummies  = sorted([c for c in feat.columns if c.startswith('marri_2_')])
genertn_dummies  = sorted([c for c in feat.columns if c.startswith('genertn_')])
fam_rela_dummies = sorted([c for c in feat.columns if c.startswith('fam_rela_')])
nominal_dummies = (tins_dummies + region_dummies + occp_dummies + house_dummies + live_t_dummies
                   + marri_2_dummies + genertn_dummies + fam_rela_dummies)

# ── Column roles for SCALING (consumed by make_scaler below) ─────────────────────
# Three groups, kept in THIS ORDER because make_scaler() addresses them as contiguous index
# ranges of the imputer's output:
#   ordinal_scaled_cols -> mode-imputed, THEN standardised  (3+ ordered levels)
#   binary_cols         -> mode-imputed, NOT standardised   (0/1 indicators, incl. one-hot)
#   continuous_cols     -> median-imputed, THEN standardised
#
# Binary indicators are left unscaled whether they came from one-hot encoding or were already
# binary in the raw data -- the distinction is meaningless for scaling, since a raw binary IS
# a one-hot column with its reference level dropped. Standardising one manufactures an extreme
# value out of nothing: a column with proportion p sends its minority rows to sqrt((1-p)/p).
# Measured on this cohort's training rows, scaling would put BS12_1 at z = 7.1 (>=50) and
# 11.5 (>=65), and BM1_0 at z = 7.2 -- purely an artefact of rarity, not signal. Left as
# recorded, a binary's maximum is always 1: "belongs to this category".
#
# The genuine ordinals below DO get standardised: they carry different ranges (1-4, 1-6, 0-5)
# so an unscaled elastic-net penalty would treat them inconsistently.
ordinal_scaled_cols = [
    # SES / demographics
    'ho_incm', 'edu', 'cfam',
    # lifestyle / behaviour
    'BP1', 'L_BR_FQ',
    # urinalysis ordinal (0-5)
    'HE_Upro',
    # dental
    'OR1',
]
binary_cols = [
    # demographics. KNHANES codes sex and town_t as 1/2 rather than 0/1; for a linear model
    # that is only an intercept shift, so they are left exactly as recorded.
    # marri_1 dropped: exact function of marri_2 (marri_1 == 2 <=> marri_2 == 0), so it is
    # perfectly collinear with the marri_2_0 dummy -- the never-married contrast is marri_2_0
    'sex', 'town_t',
    # lifestyle / behavior
    'sm_presnt', 'BD1', 'pa_aerobic', 'BS12_1',
    # dental
    'O_pain',
    'MO4_00', 'BM1_0',  # brushed yesterday (0/1), as recorded by KNHANES -- see the caveat in §4.3
    # derived
    'diabetes',
    # interdental dropped: BM2_1/BM2_3 absent from 2023-2024 files (100% missing on test year)
    # hypertension dropped (C-parallel): redundant with raw HE_sbp/HE_dbp, which are kept
    # one-hot blocks (tins, region, occp, house, live_t, marri_2, genertn, fam_rela)
] + nominal_dummies

# The IMPUTER treats both groups identically (most_frequent); only the SCALER separates them.
ordinal_cols = ordinal_scaled_cols + binary_cols
continuous_cols = [
    # demographics
    'age',
    # original lab
    'HE_BMI', 'eGFR',
    # additional lab
    'HE_Uacid', 'HE_TG', 'HE_sbp', 'HE_dbp', 'HE_chol', 'HE_HDL_st2',
    # urinalysis continuous
    'HE_Ucrea',
    # dietary
    'N_CA', 'N_NA', 'N_PROT', 'N_TDF', 'LS_VEG1', 'LS_FRUIT', 'N_SUGAR',
    # dental composite
    'brush_freq',
]
feature_cols = ordinal_cols + continuous_cols


# Temporal split: train on 2020-2023, test on 2024 (masks from df_subset; feat shares its index)
train_mask = df_subset['year'].isin([2020.0, 2021.0, 2022.0, 2023.0])
test_mask  = df_subset['year'].isin([2024.0])

# ── Minimum-count guard on one-hot dummies ───────────────────────────────────────
# A dummy with only a handful of positive cases carries no estimable coefficient, and it is
# also the worst case for feature scaling: standardising a binary column with proportion p
# maps its positive rows to sqrt((1-p)/p), i.e. z ~ 107 for a single case in 11,536. The
# scaling half of that problem is fixed by make_scaler() below; this guard removes the
# columns that were uninformative regardless.
#
# Counted on TRAIN ROWS ONLY, so the 2024 hold-out never informs the feature set.
# Respondents in a dropped level fall back into their parent's reference category -- with
# fewer than MIN_DUMMY_COUNT cases there is nothing to learn from them either way. This also
# clears the pre-existing sparse levels (fam_rela_4/7/9/10, marri_2_5), not just the *_nan
# columns introduced by dummy_na=True.
MIN_DUMMY_COUNT = 20
_dummy_counts = feat.loc[train_mask, nominal_dummies].sum()
_rare_dummies = sorted(_dummy_counts[_dummy_counts < MIN_DUMMY_COUNT].index)
if _rare_dummies:
    print(f"Dropped {len(_rare_dummies)} rare dummies (< {MIN_DUMMY_COUNT} TRAIN positives):")
    for _col in sorted(_rare_dummies, key=lambda c: -_dummy_counts[c]):
        print(f"  {_col:<16s}{int(_dummy_counts[_col]):>6,}")
    print()
    nominal_dummies = [c for c in nominal_dummies if c not in _rare_dummies]
    binary_cols     = [c for c in binary_cols     if c not in _rare_dummies]
    ordinal_cols    = [c for c in ordinal_cols    if c not in _rare_dummies]
    feature_cols    = [c for c in feature_cols    if c not in _rare_dummies]
# The three role lists must still concatenate back to ordinal_cols in order, or make_scaler's
# index ranges silently address the wrong columns.
assert ordinal_cols == ordinal_scaled_cols + binary_cols, (
    "ordinal_cols must stay ordered as [ordinal_scaled | binary] for make_scaler()")
print(f"Feature count after guard: {len(feature_cols)}  "
      f"({len(ordinal_cols)} ordinal/dummy + {len(continuous_cols)} continuous)")

X_train = feat.loc[train_mask, feature_cols].replace([np.inf, -np.inf], np.nan)
X_test  = feat.loc[test_mask,  feature_cols].replace([np.inf, -np.inf], np.nan)
y_train = df_subset.loc[train_mask, 'O_TC_']
y_test  = df_subset.loc[test_mask,  'O_TC_']

# ── Survey weights — EVALUATION ONLY ─────────────────────────────────────────────
# wt_itvex enters only where a metric is REPORTED, never where a model is fitted or
# selected. Fitting stays unweighted, the standard choice for prediction models on complex
# surveys: the learned decision rule is unchanged, but every reported discrimination and
# calibration number becomes a population estimate for Korea rather than a description of
# the raw sample.
#
# psu / kstrata are deliberately NOT used. They drive design-based VARIANCE estimation,
# which the sklearn metrics do not perform at all, and clustering the CV folds would change
# model selection rather than the reported metrics.
#
# NOT weightable, and therefore left as sample-level quantities (flagged where they occur):
#   - calibration_curve  (sklearn accepts no sample_weight)
#   - the DeLong test    (the fast algorithm assumes unweighted, i.i.d. rows)
#   - decision curve analysis (dcurves takes no weights)
#   - confusion matrices and the Youden/MCC thresholds, which are OOF/train-derived anyway
w_test = df_subset.loc[test_mask, 'wt_itvex'].astype(float)
assert w_test.index.equals(X_test.index), "test weights misaligned with X_test rows"
assert w_test.notna().all() and (w_test > 0).all(), "wt_itvex must be present and positive"
w_test_arr = w_test.to_numpy()

print()
print("── Train / Test Split " + "─" * 47)
_train_years = sorted(int(y) for y in df_subset.loc[train_mask, 'year'].unique())
_test_years  = sorted(int(y) for y in df_subset.loc[test_mask,  'year'].unique())
print(f"Train: {len(X_train):>6,} rows   years {_train_years[0]}–{_train_years[-1]}")
print(f"Test : {len(X_test):>6,} rows   years {', '.join(str(y) for y in _test_years)}")
print(f"Test-set survey weights: range {w_test_arr.min():,.0f}–{w_test_arr.max():,.0f}   "
      f"sum {w_test_arr.sum():,.0f}")

_ord_nan  = X_train[ordinal_cols].isna().sum().sum()
_cont_nan = X_train[continuous_cols].isna().sum().sum()
_tot_nan  = X_train.isna().sum().sum()
print()
print("── Missingness (train) " + "─" * 46)
print(f"Ordinal     {_ord_nan:>7,}")
print(f"Continuous  {_cont_nan:>7,}")
print(f"Total       {_tot_nan:>7,}")

def make_imputer():
    # C11: keep_empty_features=True stops SimpleImputer's default silent-drop of an
    # all-NaN column, which would otherwise shrink the transformed matrix with no error.
    return ColumnTransformer([
        ('mode',   SimpleImputer(strategy='most_frequent', keep_empty_features=True), ordinal_cols),
        ('median', SimpleImputer(strategy='median',        keep_empty_features=True), continuous_cols),
    ])

def make_scaler():
    """Standardise the ordinal and continuous blocks; pass binary indicators through as 0/1.

    Three contiguous index ranges of the imputer's output, which emits
    [ordinal_scaled_cols | binary_cols | continuous_cols] in that order:

        ordinal_scaled_cols -> StandardScaler   (3+ ordered levels, unequal ranges)
        binary_cols         -> passthrough      (0/1 indicators, incl. every one-hot dummy)
        continuous_cols     -> StandardScaler   (age, BMI, BP, dietary intakes)

    Standardising a binary manufactures an extreme value out of nothing -- a column with
    proportion p sends its minority rows to sqrt((1-p)/p), which is z ~ 107 for a single case
    in 11,536 and still z ~ 24 at the MIN_DUMMY_COUNT floor. Left as recorded, the maximum is
    always 1. Ordinals and continuous features genuinely do need standardising so the
    elastic-net penalty treats their different ranges consistently.

    ALL THREE blocks are listed explicitly and in order. A ColumnTransformer with
    remainder='passthrough' emits the TRANSFORMED block first, which would silently permute
    the columns relative to feature_cols and mislabel every SHAP row in §9. The assertions
    below lock both the width and the ordering in.
    """
    _n_ord = len(ordinal_scaled_cols)
    _n_bin = len(binary_cols)
    _n_con = len(continuous_cols)
    return ColumnTransformer([
        ('scale_ordinal', StandardScaler(), list(range(_n_ord))),
        ('passthrough',   'passthrough',    list(range(_n_ord, _n_ord + _n_bin))),
        ('scale_cont',    StandardScaler(), list(range(_n_ord + _n_bin,
                                                       _n_ord + _n_bin + _n_con))),
    ])

# C11: safety net -- if the imputer's output width ever drifts from feature_cols
# (e.g. keep_empty_features behavior changes), fail loudly here instead of a shape
# mismatch surfacing downstream (e.g. the SHAP arrays in §9, which index by feature_cols).
_imp_check = make_imputer().fit_transform(X_train)
assert _imp_check.shape[1] == len(feature_cols), (
    f"Imputer output width {_imp_check.shape[1]} != len(feature_cols) {len(feature_cols)}"
)
# The scaler must preserve the imputer's WIDTH and COLUMN ORDER: §9 indexes SHAP values by
# feature_cols against the SCALED matrix, so a reordering scaler would mislabel every row.
_sc_check = make_scaler().fit_transform(_imp_check)
_b0 = len(ordinal_scaled_cols)
_b1 = _b0 + len(binary_cols)
assert _sc_check.shape == _imp_check.shape, "scaler changed the matrix shape"
assert np.allclose(_sc_check[:, _b0:_b1], _imp_check[:, _b0:_b1]), (
    "scaler must pass the BINARY block through unchanged AND in place")
assert _sc_check[:, _b0:_b1].max() <= 2.0 + 1e-9, (
    "binary block should still be on its recorded 0/1 (or 1/2) scale after make_scaler()")
assert abs(_sc_check[:, :_b0].mean()) < 1e-9 and abs(_sc_check[:, _b1:].mean()) < 1e-9, (
    "ordinal and continuous blocks should be mean-centred after make_scaler()")

print()
print("── Scaling " + "─" * 58)
print(f"Standardised:  {len(ordinal_scaled_cols):>3} ordinal + {len(continuous_cols)} continuous")
print(f"Passthrough:   {len(binary_cols):>3} binary/one-hot (kept on 0/1 scale, not z-scored)")
del _imp_check, _sc_check, _b0, _b1
```

<h3 style="color:#87CEEB";> 6.1 Development vs. Evaluation Comparison (TRIPOD+AI Item 20c) </h3>

Table 1 (§5.1–5.3) splits by **outcome**. Item 20c instead requires a split by **wave** so a
reviewer can confirm the 2024 evaluation set is distributionally comparable to the 2020–2023
development set. The COVID-affected **2020** wave is isolated as its own column, with a dedicated
`SMD(2020 vs 2021–2023)` measuring whether it distorts the development set.

**Standardized mean differences (SMD)** are reported instead of p-values: with n in the thousands,
ordinary t / χ² tests flag trivial differences as "significant." Convention: **|SMD| > 0.1** marks a
potentially meaningful imbalance (flagged `*`). Nominal variables use the Yang & Dalton (2012)
multinomial SMD (the `tableone` method).

```python
# ── §6.1 Development vs. Evaluation Comparison — TRIPOD+AI Item 20c ────────────
# Splits df_subset (same raw source as the outcome-split Table 1 above) by survey
# WAVE. 2020 is isolated because it is the COVID-affected wave. Two SMD columns:
#   SMD(2020 vs 2021-23)  -> does the COVID wave distort the development set?
#   SMD(Dev 2020-23 vs 2024) -> is the evaluation set comparable to development?
# |SMD| > 0.1 = potentially meaningful imbalance (flagged *).

# wave masks
_m2020 = df_subset["year"] == 2020.0
_m2123 = df_subset["year"].isin([2021.0, 2022.0, 2023.0])
_m2024 = df_subset["year"] == 2024.0
_mdev  = _m2020 | _m2123                       # development set (2020-2023)

def _smd_cont(a, b):
    a, b = a.dropna(), b.dropna()
    if len(a) < 2 or len(b) < 2:
        return np.nan
    pooled = np.sqrt((a.std(ddof=1) ** 2 + b.std(ddof=1) ** 2) / 2)
    return np.nan if pooled == 0 else (a.mean() - b.mean()) / pooled

def _smd_bin(a, b, pos):
    a, b = a.dropna(), b.dropna()
    if len(a) == 0 or len(b) == 0:
        return np.nan
    p1, p2 = (a == pos).mean(), (b == pos).mean()
    pooled = np.sqrt((p1 * (1 - p1) + p2 * (1 - p2)) / 2)
    return np.nan if pooled == 0 else (p1 - p2) / pooled

def _smd_multi(a, b):
    # Yang & Dalton (2012) multinomial SMD — same method as the tableone package.
    a, b = a.dropna(), b.dropna()
    if len(a) == 0 or len(b) == 0:
        return np.nan
    levels = sorted(set(a.unique()) | set(b.unique()))
    if len(levels) < 2:
        return np.nan
    pa = np.array([(a == L).mean() for L in levels])[:-1]   # drop last (redundant)
    pb = np.array([(b == L).mean() for L in levels])[:-1]
    def _cov(p):
        S = -np.outer(p, p)
        np.fill_diagonal(S, p * (1 - p))
        return S
    S = (_cov(pa) + _cov(pb)) / 2
    diff = pa - pb
    val = diff @ np.linalg.pinv(S) @ diff
    return np.sqrt(max(val, 0.0))

def _flag(s):
    return "*" if (pd.notna(s) and abs(s) > 0.1) else ""

def _n(mask):
    return int(mask.sum())

# variable dictionaries — mirror §5.1 / §5.3 (Table 1). Guarded on column existence.
_cont_vars = {
    "age": "Age (years)", "HE_HbA1c": "HbA1c (%)", "eGFR": "eGFR (mL/min/1.73m²)",
    "HE_BMI": "BMI (kg/m²)", "HE_Uacid": "Uric Acid (mg/dL)", "HE_sbp": "Systolic BP (mmHg)",
    "HE_dbp": "Diastolic BP (mmHg)", "HE_TG": "Triglycerides (mg/dL)", "brush_freq": "Brushing Frequency",
}
_bin_vars = {
    "O_TC_": ("Tooth Loss <20 (Yes)", 1),           # OUTCOME
    "sex": ("Sex (Female)", 2), "diabetes": ("Diabetes (Yes)", 1.0),
    "hypertension": ("Hypertension (Yes)", 1.0), "sm_presnt": ("Current Smoker (Yes)", 1),
    "BD1": ("Alcohol Drinker (Yes)", 2),  # 1 = NEVER drank -> positive category is 2
    "pa_aerobic": ("Aerobic PA (Yes)", 1),
    "BS12_1": ("Secondhand Smoke (Yes)", 1), "O_pain": ("Dental Pain (Yes)", 1),
    "MO4_00": ("Dental Visit past yr (Yes)", 1), "BM1_0": ("Brushed Yesterday (Yes)", 1),
    "town_t": ("Urban Area", 1),
}
_nom_vars = {
    "marri_1": "Marital Status", "occp": "Occupation", "cfam": "Household Composition",
    "live_t": "Living Arrangement", "house": "Housing Type", "tins": "Insurance Type",
    "region": "Region",
}

_hdr_2020 = f"2020·COVID (n={_n(_m2020)})"
_hdr_2123 = f"2021–2023 (n={_n(_m2123)})"
_hdr_2024 = f"2024·Eval (n={_n(_m2024)})"

# ── Continuous — mean ± SD ────────────────────────────────────────────────────
_rows_c = []
for col, label in _cont_vars.items():
    if col not in df_subset.columns:
        continue
    s = df_subset[col]
    a20, a21, a24, adev = s[_m2020].dropna(), s[_m2123].dropna(), s[_m2024].dropna(), s[_mdev].dropna()
    smd_covid = _smd_cont(a20, a21)
    smd_de    = _smd_cont(adev, a24)
    _rows_c.append({
        "Variable": label,
        _hdr_2020: f"{a20.mean():.2f} ± {a20.std():.2f}",
        _hdr_2123: f"{a21.mean():.2f} ± {a21.std():.2f}",
        _hdr_2024: f"{a24.mean():.2f} ± {a24.std():.2f}",
        "SMD 2020vs21–23": f"{smd_covid:.3f}{_flag(smd_covid)}",
        "SMD Dev vs Eval": f"{smd_de:.3f}{_flag(smd_de)}",
    })

# ── Binary + outcome — n (%) ──────────────────────────────────────────────────
_rows_b = []
for col, (label, pos) in _bin_vars.items():
    if col not in df_subset.columns:
        continue
    s = df_subset[col]
    a20, a21, a24, adev = s[_m2020].dropna(), s[_m2123].dropna(), s[_m2024].dropna(), s[_mdev].dropna()
    def _pct(x):
        return f"{int((x == pos).sum())} ({100 * (x == pos).mean():.1f}%)"
    smd_covid = _smd_bin(a20, a21, pos)
    smd_de    = _smd_bin(adev, a24, pos)
    _rows_b.append({
        "Variable": label,
        _hdr_2020: _pct(a20), _hdr_2123: _pct(a21), _hdr_2024: _pct(a24),
        "SMD 2020vs21–23": f"{smd_covid:.3f}{_flag(smd_covid)}",
        "SMD Dev vs Eval": f"{smd_de:.3f}{_flag(smd_de)}",
    })

# ── Nominal — multinomial SMD (Yang & Dalton) ─────────────────────────────────
_rows_n = []
for col, label in _nom_vars.items():
    if col not in df_subset.columns:
        continue
    s = df_subset[col]
    smd_covid = _smd_multi(s[_m2020], s[_m2123])
    smd_de    = _smd_multi(s[_mdev], s[_m2024])
    _rows_n.append({
        "Variable": label, "n levels": int(s.dropna().nunique()),
        "n 2020": _n(_m2020 & s.notna()), "n 21–23": _n(_m2123 & s.notna()), "n 2024": _n(_m2024 & s.notna()),
        "SMD 2020vs21–23": f"{smd_covid:.3f}{_flag(smd_covid)}",
        "SMD Dev vs Eval": f"{smd_de:.3f}{_flag(smd_de)}",
    })

print("══ §6.1 Development (2020–2023) vs Evaluation (2024) — TRIPOD+AI Item 20c ══\n")
print("── Continuous (mean ± SD) ──────────────────────────────────────────────")
print(pd.DataFrame(_rows_c).to_string(index=False))
print("\n── Binary + Outcome (n, %) ─────────────────────────────────────────────")
print(pd.DataFrame(_rows_b).to_string(index=False))
print("\n── Nominal (multinomial SMD) ───────────────────────────────────────────")
print(pd.DataFrame(_rows_n).to_string(index=False))

# ── Imbalance summary ─────────────────────────────────────────────────────────
def _vals(rows, key):
    out = []
    for r in rows:
        try:
            out.append(abs(float(r[key].rstrip("*"))))
        except (ValueError, KeyError):
            pass
    return out
_all = _rows_c + _rows_b + _rows_n
_de   = [v for v in _vals(_all, "SMD Dev vs Eval") if pd.notna(v)]
_cov  = [v for v in _vals(_all, "SMD 2020vs21–23") if pd.notna(v)]
print("\n── Summary (|SMD| > 0.1 = imbalanced) ──────────────────────────────────")
print(f"Dev vs Eval : {sum(v > 0.1 for v in _de)}/{len(_de)} variables imbalanced "
      f"(max |SMD| = {max(_de):.3f})")
print(f"2020 vs 21–23: {sum(v > 0.1 for v in _cov)}/{len(_cov)} variables imbalanced "
      f"(max |SMD| = {max(_cov):.3f})  ← COVID-distortion check")
```

<h2 style="color:#FFA500";> 8. Hyperparameter Tuning — Optuna </h2>

**Method:** Single **50-trial Optuna TPE pass** per model using 5-fold `StratifiedKFold`, scored by `average_precision`. Class-imbalance handling is fixed for every model, not searched — see the note below the table.

**Standardization policy:** `StandardScaler` is added to the Logistic Regression pipeline only — tree-based models are scale-invariant by construction.

| Model | Key hyperparameters tuned |
|-------|--------------------------|
| **Random Forest** | `n_estimators`, `max_depth`*, `min_samples_leaf`* |
| **XGBoost** | `n_estimators`, `max_depth`, `learning_rate`, `subsample`, `colsample_bytree`, `min_child_weight`, `gamma`, `reg_lambda`* |
| **LightGBM** | `n_estimators`, `num_leaves`, `learning_rate`, `min_child_samples`, `reg_lambda`* |
| **CatBoost** | `iterations`, `depth`, `learning_rate`, `l2_leaf_reg`* |
| **Logistic Regression** | `C`* (inverse strength), `l1_ratio` (ElasticNet mix, SAGA solver) |

\* = regularization parameter

**`scale_pos_weight` is pinned, not tuned.** XGBoost's imbalance parameter is fixed to `n_neg/n_pos` in the objective function below, not left for Optuna to search — it was previously tuned over 1.0–5.0, which handed XGBoost one degree of freedom no other model had. The other four models handle class imbalance the same fixed way: `class_weight="balanced"` (Random Forest, LightGBM, Logistic Regression) or `auto_class_weights="Balanced"` (CatBoost), both of which resolve to the same `n_neg/n_pos` ratio — none of them get to search over it either. If `scale_pos_weight` were free to roam in Optuna's search space, XGBoost would gain an extra tunable knob for imbalance handling that no other model has: it could win the comparison not because gradient boosting is a better architecture, but because it got to hill-climb on the AUPRC objective using a parameter that acts almost like fitting on the evaluation metric itself. Pinning it puts all five models on equal footing for imbalance correction, isolating the comparison to genuine model/architecture differences — the right call for a model-comparison study, not a stylistic one.

**Method:** Optuna TPE search (sample-efficient — learns from past trials, unlike random search; 50 trials per model, `scoring = 'average_precision'`)

| Model                   | Key Hyperparameters                                                                                                              | Search Range                                                                                                       |
| :---------------------- | :------------------------------------------------------------------------------------------------------------------------------- | :----------------------------------------------------------------------------------------------------------------- |
| **Random Forest**       | `n_estimators`, `max_depth`, `min_samples_leaf`                                                                                  | 100–500 (step 100) · {5,10,15,None} · 1–10                                                                        |
| **XGBoost**             | `n_estimators`, `max_depth`, `learning_rate`, `subsample`, `colsample_bytree`, `min_child_weight`, `gamma`, `reg_lambda` | 100–700 · 3–7 · 0.01–0.3 (log) · 0.7–1.0 · 0.5–1.0 · 1–6 · 0–5 · 0.01–10 (log) |
| **LightGBM**            | `n_estimators`, `num_leaves`, `learning_rate`, `min_child_samples`, `reg_lambda`                                                 | 100–500 · 15–63 · 0.01–0.3 (log) · 5–20 · 0.01–10 (log)                                                          |
| **CatBoost**            | `iterations`, `depth`, `learning_rate`, `l2_leaf_reg`                                                                           | 100–500 · 3–8 · 0.01–0.3 (log) · 1–10                                                                             |
| **Logistic Regression** | `C`, `l1_ratio` (ElasticNet via SAGA)                                                                                           | 0.001–100 (log) · 0.0–1.0                                                                                         |

`scale_pos_weight` (XGBoost's imbalance parameter) is pinned to `n_neg/n_pos`, not tuned — see §8.

<h3 style="color:yellow";> 8.1 Objective Function Definitions </h3>

Define one Optuna objective per model. Each objective runs 5-fold `StratifiedKFold` CV and returns the mean `average_precision`. The imputer is rebuilt inside every CV fold to prevent data leakage.

**Regularization parameters per model:**

| Model | Regularization parameter(s) | Type | Search range |
|---|---|---|---|
| Random Forest | `max_depth`, `min_samples_leaf` | Structural (limits tree size) | depth ∈ {5,10,15,None} · leaf ∈ [1,10] |
| XGBoost | `reg_lambda` | L2 weight penalty | [0.01, 10] log-scale |
| LightGBM | `reg_lambda` | L2 weight penalty | [0.01, 10] log-scale |
| CatBoost | `l2_leaf_reg` | L2 leaf penalty | [1, 10] |
| Logistic Regression | `C`, `l1_ratio` | Inverse ElasticNet strength; L1/L2 mix ratio | C ∈ [0.001, 100] log-scale · l1_ratio ∈ [0, 1] |

```python
import optuna
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, average_precision_score
from sklearn.pipeline import Pipeline
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

optuna.logging.set_verbosity(optuna.logging.WARNING)

cv       = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
N_TRIALS = 50

# Class weighting: RF / LightGBM / LogisticRegression use class_weight='balanced' and
# CatBoost uses auto_class_weights='Balanced' -- both are STRINGS that sklearn/CatBoost
# resolve to n_neg/n_pos INTERNALLY, at whatever data is passed to .fit(). Under CV
# (cross_val_score / cross_val_predict clone the estimator and fit each clone on its own
# fold), that means the ratio is recomputed fresh from each fold's OWN training partition,
# not from the full X_train.
#
# XGBoost has no such string option -- scale_pos_weight is just a number, so it can't be
# resolved per fold by passing a flag into the constructor. SCALE_POS_WEIGHT below is the
# full-training-set ratio, kept ONLY for the final refit in §9.1: fitting once on all of
# X_train has no folds to be inconsistent across, so the full-set ratio there is exactly
# what 'balanced' would also give. It is NOT used during tuning below -- obj_xgb instead
# recomputes the ratio inside a manual per-fold loop, so XGBoost's weighting is fold-local
# during CV just like the other four models, instead of one frozen value reused everywhere.
SCALE_POS_WEIGHT = float((y_train == 0).sum() / (y_train == 1).sum())
print(f"Full-training-set scale_pos_weight = {SCALE_POS_WEIGHT:.4f}"
      f"  (used only for the final refit in §9.1 -- tuning below is fold-local)")

def cv_score(estimator):
    return cross_val_score(estimator, X_train, y_train,
                           cv=cv, scoring='average_precision', n_jobs=1).mean()   # n_jobs=1: avoid macOS fork-kill under memory pressure

def obj_rf(trial):
    model = Pipeline([
        ('imputer', make_imputer()),
        ('model', RandomForestClassifier(
            n_estimators     = trial.suggest_int('n_estimators', 100, 500, step=100),
            max_depth        = trial.suggest_categorical('max_depth', [5, 10, 15, None]),
            min_samples_leaf = trial.suggest_int('min_samples_leaf', 1, 10),
            class_weight='balanced', random_state=42, n_jobs=-1))
    ])
    return cv_score(model)

def obj_xgb(trial):
    # scale_pos_weight has no 'balanced' string like sklearn/CatBoost, so it cannot be
    # resolved per fold by cross_val_score the way the other four objectives are -- whatever
    # value is baked into the constructor stays frozen across every fold it clones. To match
    # their fold-local behavior instead, this objective manually loops over the SAME
    # StratifiedKFold splits `cv` would use, recomputing n_neg/n_pos from each fold's OWN
    # training partition only -- exactly what class_weight='balanced' does internally for
    # the other models.
    params = dict(
        n_estimators      = trial.suggest_int('n_estimators', 100, 700, step=100),
        max_depth         = trial.suggest_int('max_depth', 3, 7),
        learning_rate     = trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        subsample         = trial.suggest_float('subsample', 0.7, 1.0),
        colsample_bytree  = trial.suggest_float('colsample_bytree', 0.5, 1.0),
        min_child_weight  = trial.suggest_int('min_child_weight', 1, 6),
        gamma             = trial.suggest_float('gamma', 0.0, 5.0),
        reg_lambda        = trial.suggest_float('reg_lambda', 1e-2, 10.0, log=True),
    )
    fold_scores = []
    for train_idx, val_idx in cv.split(X_train, y_train):
        # .iloc, not .loc: StratifiedKFold.split returns POSITIONAL indices into whatever
        # was passed to it, not the DataFrame's own row labels.
        X_tr, X_val = X_train.iloc[train_idx], X_train.iloc[val_idx]
        y_tr, y_val = y_train.iloc[train_idx], y_train.iloc[val_idx]
        fold_spw = float((y_tr == 0).sum() / (y_tr == 1).sum())
        fold_model = Pipeline([
            ('imputer', make_imputer()),
            ('model', XGBClassifier(
                **params,
                scale_pos_weight=fold_spw,   # fold-local, not the global SCALE_POS_WEIGHT
                eval_metric='logloss', random_state=42, nthread=1))
        ])
        fold_model.fit(X_tr, y_tr)
        fold_proba = fold_model.predict_proba(X_val)[:, 1]
        fold_scores.append(average_precision_score(y_val, fold_proba))
    return float(np.mean(fold_scores))

def obj_lgbm(trial):
    model = Pipeline([
        ('imputer', make_imputer()),
        ('model', LGBMClassifier(
            n_estimators      = trial.suggest_int('n_estimators', 100, 500, step=100),
            num_leaves        = trial.suggest_int('num_leaves', 15, 63),
            learning_rate     = trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
            min_child_samples = trial.suggest_int('min_child_samples', 5, 20),
            reg_lambda        = trial.suggest_float('reg_lambda', 1e-2, 10.0, log=True),
            class_weight='balanced', verbose=-1, random_state=42, n_jobs=1))
    ])
    return cv_score(model)

def obj_cat(trial):
    model = Pipeline([
        ('imputer', make_imputer()),
        ('model', CatBoostClassifier(
            iterations    = trial.suggest_int('iterations', 100, 500, step=100),
            depth         = trial.suggest_int('depth', 3, 8),
            learning_rate = trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
            l2_leaf_reg   = trial.suggest_float('l2_leaf_reg', 1.0, 10.0),
            auto_class_weights='Balanced', random_seed=42, verbose=0))
    ])
    return cv_score(model)

def obj_lr(trial):
    model = Pipeline([
        ('imputer', make_imputer()),
        ('scaler', make_scaler()),   # continuous only -- dummies stay 0/1, see §6
        ('model', LogisticRegression(
            # `penalty` was deprecated in scikit-learn 1.8 and is REMOVED in 1.10. The
            # penalty is now inferred from l1_ratio (0 = L2, 1 = L1, in between = elastic
            # net), so passing l1_ratio alone is the forward-compatible spelling. Verified
            # to give coefficients identical to the old penalty='elasticnet' on 1.8.
            C            = trial.suggest_float('C', 1e-3, 1e2, log=True),
            l1_ratio     = trial.suggest_float('l1_ratio', 0.0, 1.0),
            solver='saga', class_weight='balanced', max_iter=1000, random_state=42))
    ])
    return cv_score(model)

objectives = {
    'Random Forest':       obj_rf,
    'XGBoost':             obj_xgb,
    'LightGBM':            obj_lgbm,
    'CatBoost':            obj_cat,
    'Logistic Regression': obj_lr,
}
```

<h3 style="color:yellow";> 8.2 Tuning — 50-Trial Optuna Pass </h3>

50-trial Optuna TPE study per model using 5-fold CV and the search spaces defined in §8.1. Best params and best CV AUPRC are stored in `results` for the refit and test-set evaluation in §9.

```python
results = {}
for name, objective in objectives.items():
    study = optuna.create_study(
        direction='maximize',
        sampler=optuna.samplers.TPESampler(seed=42))
    study.optimize(objective, n_trials=N_TRIALS, show_progress_bar=False)
    results[name] = {
        'best_params':   study.best_params,
        'best_cv_score': study.best_value,
        'study':         study
    }
    print(f"{name}: Best AUPRC = {study.best_value:.4f} | Params: {study.best_params}")

results_df = pd.DataFrame(
    {name: {'AUPRC': v['best_cv_score']} for name, v in results.items()}
).T.sort_values('AUPRC', ascending=False)
print(results_df)
```

<h2 style="color:#FFA500";> 9. Refit & Evaluate Tuned Models on the Test Set </h2>

Rebuild each model from its Optuna-best params, fit on the 2020–2023 training fold, and score on the held-out 2024 test set. This compares cross-validated AUPRC (Optuna objective) against out-of-sample test AUC and AUPRC, selecting the best model by **OOF AUPRC** — out-of-fold on the 2020–2023 training years only — for the diagnostics in §10. Test AUC / AUPRC are reported but never used to select; see the rationale in the code cell.

> Replaces the earlier PyCaret `compare_models` screening — Optuna already tunes the chosen models directly, so a separate AutoML pass is redundant.

**Note — predictive metrics are unweighted.** The AUC / AUPRC / confusion matrices below are computed on the raw sample, *not* survey-weighted. KNHANES sampling weights are applied only in the inferential (odds-ratio) model. Rank metrics (AUC/AUPRC) are largely insensitive to weighting, but all reported predictive performance describes the **sample**, not the Korean population.

<h3 style="color:yellow";> 9.1 Refit Tuned Models & Select Best by OOF AUPRC </h3>

```python
from sklearn.metrics import roc_auc_score, average_precision_score
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_val_predict

def build_estimator(name, params):
    if name == "Random Forest":
        return Pipeline([
            ("imputer", make_imputer()),
            ("model", RandomForestClassifier(
                class_weight="balanced", random_state=42, n_jobs=-1, **params))
        ])
    if name == "XGBoost":
        # scale_pos_weight is PINNED in §8 and no longer tuned, so it should not appear in
        # best_params. Strip it defensively: a `results` dict left over from a session where
        # it WAS tuned would otherwise raise "got multiple values for keyword argument
        # 'scale_pos_weight'" on a partial re-run of this cell.
        params = {k: v for k, v in params.items() if k != "scale_pos_weight"}
        return Pipeline([
            ("imputer", make_imputer()),
            ("model", XGBClassifier(
                scale_pos_weight=SCALE_POS_WEIGHT,
                eval_metric="logloss", random_state=42, nthread=1, **params))
        ])
    if name == "LightGBM":
        return Pipeline([
            ("imputer", make_imputer()),
            ("model", LGBMClassifier(
                class_weight="balanced", verbose=-1, random_state=42, n_jobs=1, **params))
        ])
    if name == "CatBoost":
        return Pipeline([
            ("imputer", make_imputer()),
            ("model", CatBoostClassifier(
                auto_class_weights="Balanced", random_seed=42, verbose=0, **params))
        ])
    if name == "Logistic Regression":
        return Pipeline([
            ("imputer", make_imputer()),
            ("scaler", make_scaler()),   # continuous only -- dummies stay 0/1, see §6
            # penalty inferred from l1_ratio (arrives via **params) -- see §8. The explicit
            # penalty= argument is removed in scikit-learn 1.10.
            ("model", LogisticRegression(
                class_weight="balanced",
                solver="saga", max_iter=1000, random_state=42, **params))
        ])
    raise ValueError(name)

def _xgb_oof_proba(params):
    # Mirrors §8's obj_xgb: cross_val_predict would clone build_estimator's XGBoost pipeline
    # (frozen at the full-training-set SCALE_POS_WEIGHT) onto every fold, so XGBoost's OOF
    # predictions would be the only ones NOT using a fold-local class-balance ratio -- RF /
    # LightGBM / LR resolve 'balanced' and CatBoost resolves 'Balanced' fresh inside each
    # fold cross_val_predict fits. This loop gives XGBoost the same fold-local treatment.
    _params = {k: v for k, v in params.items() if k != "scale_pos_weight"}
    oof = np.zeros(len(X_train))
    for train_idx, val_idx in cv.split(X_train, y_train):
        X_tr, X_val = X_train.iloc[train_idx], X_train.iloc[val_idx]
        y_tr        = y_train.iloc[train_idx]
        fold_spw    = float((y_tr == 0).sum() / (y_tr == 1).sum())
        fold_model = Pipeline([
            ("imputer", make_imputer()),
            ("model", XGBClassifier(
                scale_pos_weight=fold_spw,
                eval_metric="logloss", random_state=42, nthread=1, **_params))
        ])
        fold_model.fit(X_tr, y_tr)
        oof[val_idx] = fold_model.predict_proba(X_val)[:, 1]
    return oof

fitted_models = {}
oof_proba_by_model = {}
rows = []
for name, info in results.items():
    est = build_estimator(name, info["best_params"])
    if name == "XGBoost":
        _oof_proba = _xgb_oof_proba(info["best_params"])
    else:
        _oof_proba = cross_val_predict(est, X_train, y_train, cv=cv, method="predict_proba")[:, 1]
    oof_proba_by_model[name] = _oof_proba
    oof_auprc  = average_precision_score(y_train, _oof_proba)
    est.fit(X_train, y_train)
    fitted_models[name] = est
    proba = est.predict_proba(X_test)[:, 1]
    # Test metrics are SURVEY-WEIGHTED (population estimates). CV/OOF above stay unweighted:
    # they drive model SELECTION, an internal choice made on the development sample.
    test_auc   = roc_auc_score(y_test, proba, sample_weight=w_test_arr)
    test_auprc = average_precision_score(y_test, proba, sample_weight=w_test_arr)
    rows.append({"Model": name,
                 "CV AUPRC":     info["best_cv_score"],
                 "OOF AUPRC":  oof_auprc,
                 "Test AUC":   test_auc,
                 "Test AUPRC": test_auprc})

# Rank by OOF AUPRC -- out-of-fold predictions on the 2020-2023 TRAINING data only.
#
# Selecting on Test AUPRC (the previous behaviour) makes every reported 2024 number a
# maximum over 5 candidate models, so the hold-out stops functioning as a hold-out and
# the optimism propagates into everything keyed on best_name: SHAP, top_shap_vars, the
# SHAP-informed regression in §9, the Youden/MCC thresholds, DCA and calibration. The
# bootstrap CIs in §9 do not cover that selection bias.
#
# OOF AUPRC is also the better choice over Optuna's own best_cv_score ("CV AUPRC"),
# which is the maximum over 50 trials and therefore optimistically biased; the OOF
# column re-scores the single chosen configuration on held-out folds.
#
# Test AUPRC / Test AUC remain in the table -- reported, never used to choose.
eval_df = (pd.DataFrame(rows)
             .set_index("Model")
             .sort_values("OOF AUPRC", ascending=False))
print(eval_df.round(4).to_string())
print("CAPTION: 'CV AUPRC' and 'OOF AUPRC' are UNWEIGHTED -- selection metrics computed on"
      "\nthe 2020-2023 development sample. 'Test AUC' and 'Test AUPRC' are SURVEY-WEIGHTED"
      "\npopulation estimates on the 2024 hold-out. Compare down a column, not across a row."
      "\nSelection is unweighted by design, so the selected model need not have the best"
      "\nweighted test score -- that is the hold-out working, not an inconsistency.")

best_name  = eval_df.index[0]  # selected by OOF AUPRC -- leakage-free, never sees 2024
best_model = fitted_models[best_name]
print(f'Best model by OOF AUPRC: {best_name} (OOF AUPRC = {eval_df.loc[best_name, "OOF AUPRC"]:.4f})'
      f' | Test AUPRC: {eval_df.loc[best_name, "Test AUPRC"]:.4f}'
      f' | Test AUC: {eval_df.loc[best_name, "Test AUC"]:.4f}'
      f' | Optuna CV AUPRC: {eval_df.loc[best_name, "CV AUPRC"]:.4f}')
```

<h3 style="color:yellow";> 9.2 Bootstrap 95% CIs — Test-Set Metrics </h3>

```python
# ── Test-set metrics with bootstrap 95% CIs — all tuned models ───────────────
# AUC (ROC), AUPRC (average precision) and Brier score on the 2024 hold-out test
# set, each with a 95% CI from a stratified (prevalence-preserving) bootstrap of
# the test rows. 2000 resamples, fixed seed → fully reproducible on "Run All".
# Depends on fitted_models / X_test / y_test from the model-evaluation cell (§9).
from sklearn.metrics import roc_auc_score, average_precision_score, brier_score_loss

N_BOOT_CI = 2000
_yte = y_test.to_numpy()
_wte = w_test.to_numpy()      # survey weights, applied to every metric below
_idx0 = np.where(_yte == 0)[0]
_idx1 = np.where(_yte == 1)[0]

# Draw the bootstrap resamples ONCE so every model is scored on the same rows (paired)
_rng_ci = np.random.default_rng(42)
_boot_idx = [
    np.concatenate([_rng_ci.choice(_idx0, _idx0.size, replace=True),
                    _rng_ci.choice(_idx1, _idx1.size, replace=True)])
    for _ in range(N_BOOT_CI)
]

def _boot_ci(proba, metric):
    # weights are indexed alongside y/proba so each resample carries its own rows' weights
    vals = np.fromiter((metric(_yte[bi], proba[bi], sample_weight=_wte[bi]) for bi in _boot_idx),
                       dtype=float, count=N_BOOT_CI)
    lo, hi = np.percentile(vals, [2.5, 97.5])
    return lo, hi

_rows_ci = []
for _name, _model in fitted_models.items():
    _p = _model.predict_proba(X_test)[:, 1]
    _pt = {"AUC":   roc_auc_score(_yte, _p, sample_weight=_wte),
           "AUPRC": average_precision_score(_yte, _p, sample_weight=_wte),
           "Brier": brier_score_loss(_yte, _p, sample_weight=_wte)}
    _ci = {"AUC":   _boot_ci(_p, roc_auc_score),
           "AUPRC": _boot_ci(_p, average_precision_score),
           "Brier": _boot_ci(_p, brier_score_loss)}
    _rows_ci.append({"Model": _name,
                     **{f"{m} (95% CI)": f"{_pt[m]:.3f} ({_ci[m][0]:.3f}–{_ci[m][1]:.3f})"
                        for m in ("AUC", "AUPRC", "Brier")}})

metrics_ci_df = pd.DataFrame(_rows_ci).set_index("Model")
print(f"Test-set metrics with bootstrap 95% CIs "
      f"(n_test={len(_yte)}, {N_BOOT_CI} stratified resamples, seed=42)")
print(metrics_ci_df.to_string())
```

<h3 style="color:#87CEEB";> 9.3 Sex-Stratified Discrimination — AUROC / AUPRC by Sex </h3>

```python
# ── §9.3 Sex-stratified discrimination — AUROC / AUPRC by sex, all tuned models ──
# Reviewer request: report discrimination separately for male and female respondents
# within this cohort, to check the models perform equally well across sexes. Reuses the
# fitted pipelines and the 2024 hold-out test set from §9; CIs use the same stratified
# bootstrap recipe as the all-models table above (2000 resamples, seed 42), resampled
# within each sex subset so the interval reflects that subset's size.
from sklearn.metrics import roc_auc_score, average_precision_score

_yte = y_test.to_numpy()  # self-contained: re-derive rather than lean on §9 leakage
_wte = w_test.to_numpy()  # survey weights (evaluation only) -- see the split cell in §6

# Sex aligned to the test rows. KNHANES codes sex 1=male / 2=female and the pipeline
# never recodes it; feat shares df_subset's index, so test_mask selects the same rows
# (in the same order) as X_test — asserted below before use.
_sex_test = df_subset.loc[test_mask, 'sex'].map({1: 'Male', 2: 'Female'})
assert _sex_test.index.equals(X_test.index), "sex vector misaligned with X_test rows"
_sex_arr = _sex_test.to_numpy()

N_BOOT_SEX = 2000
_rng_sex = np.random.default_rng(42)

# Per-sex, prevalence-preserving (stratified) bootstrap index pools over the test set,
# mirroring §9's _boot_idx. Resample the outcome-0 and outcome-1 rows separately so both
# classes are always present; None flags a subset that is single-class (metric undefined).
_sex_boot = {}
for _s in ('Male', 'Female'):
    _rows_s = np.where(_sex_arr == _s)[0]
    _y_s = _yte[_rows_s]
    _i0, _i1 = _rows_s[_y_s == 0], _rows_s[_y_s == 1]
    _sex_boot[_s] = {
        'rows': _rows_s,
        'boot': ([np.concatenate([_rng_sex.choice(_i0, _i0.size, replace=True),
                                  _rng_sex.choice(_i1, _i1.size, replace=True)])
                  for _ in range(N_BOOT_SEX)] if (_i0.size and _i1.size) else None),
    }

def _sex_ci(proba, boot, metric):
    if boot is None:
        return None
    vals = np.fromiter((metric(_yte[bi], proba[bi], sample_weight=_wte[bi]) for bi in boot),
                       dtype=float, count=N_BOOT_SEX)
    lo, hi = np.percentile(vals, [2.5, 97.5])
    return lo, hi

def _fmt(pt, ci):
    if np.isnan(pt):
        return "n/a (single-class subset)"
    return f"{pt:.3f} ({ci[0]:.3f}–{ci[1]:.3f})" if ci else f"{pt:.3f} (CI n/a)"

_rows_sex = []
for _name, _model in fitted_models.items():
    _p_full = _model.predict_proba(X_test)[:, 1]
    for _s in ('Male', 'Female'):
        _rows_s, _boot_s = _sex_boot[_s]['rows'], _sex_boot[_s]['boot']
        _y_s, _p_s, _w_s = _yte[_rows_s], _p_full[_rows_s], _wte[_rows_s]
        _one_class = len(np.unique(_y_s)) < 2
        _auc = np.nan if _one_class else roc_auc_score(_y_s, _p_s, sample_weight=_w_s)
        _ap = np.nan if _one_class else average_precision_score(_y_s, _p_s, sample_weight=_w_s)
        _rows_sex.append({
            "Model": _name, "Sex": _s,
            "n": int(_rows_s.size), "n_pos": int(_y_s.sum()),
            "AUROC (95% CI)": _fmt(_auc, _sex_ci(_p_full, _boot_s, roc_auc_score)),
            "AUPRC (95% CI)": _fmt(_ap, _sex_ci(_p_full, _boot_s, average_precision_score)),
        })

sex_metrics_df = pd.DataFrame(_rows_sex).set_index(["Model", "Sex"])
print(f"Sex-stratified test-set discrimination "
      f"(age≥{age_threshold} cohort, n_test={len(_yte)}, "
      f"{N_BOOT_SEX} stratified resamples, seed=42)")
print(sex_metrics_df.to_string())
```

<h3 style="color:#87CEEB";> 9.4 DeLong Significance Test — Pairwise Model AUC Comparison </h3>

```python
# ── DeLong test — pairwise significance of Test AUC differences (Sun & Xu 2014 fast algorithm) ──
# Test AUPRC gaps between models are often within noise on a single temporal holdout; DeLong gives
# a p-value for whether the Test AUC gap between two models is real, using the fact that both models
# are scored on the same test rows (paired), not independent samples.
from scipy import stats

def _compute_midrank(x):
    J = np.argsort(x)
    Z = x[J]
    N = len(x)
    T = np.zeros(N, dtype=float)
    i = 0
    while i < N:
        j = i
        while j < N and Z[j] == Z[i]:
            j += 1
        T[i:j] = 0.5 * (i + j - 1) + 1
        i = j
    T2 = np.empty(N, dtype=float)
    T2[J] = T
    return T2

def _fast_delong(preds, label_1_count):
    m = label_1_count
    n = preds.shape[1] - m
    positive = preds[:, :m]
    negative = preds[:, m:]
    k = preds.shape[0]

    tx = np.empty([k, m], dtype=float)
    ty = np.empty([k, n], dtype=float)
    tz = np.empty([k, m + n], dtype=float)
    for r in range(k):
        tx[r, :] = _compute_midrank(positive[r, :])
        ty[r, :] = _compute_midrank(negative[r, :])
        tz[r, :] = _compute_midrank(preds[r, :])
    aucs = tz[:, :m].sum(axis=1) / m / n - float(m + 1) / (2 * n)
    v01 = (tz[:, :m] - tx[:, :]) / n
    v10 = 1.0 - (tz[:, m:] - ty[:, :]) / m
    sx = np.cov(v01)
    sy = np.cov(v10)
    delongcov = sx / m + sy / n
    return aucs, delongcov

def _calc_pvalue(aucs, sigma):
    l = np.array([[1, -1]])
    z = np.abs(np.diff(aucs)) / np.sqrt(l.dot(sigma).dot(l.T))
    return 2 * (1 - stats.norm.cdf(z))[0, 0]

def delong_roc_test(y_true, prob_a, prob_b):
    order = np.argsort(-y_true)
    y_sorted = y_true[order]
    m = int(np.sum(y_sorted == 1))
    preds = np.vstack([prob_a[order], prob_b[order]])
    aucs, delongcov = _fast_delong(preds, m)
    return aucs, _calc_pvalue(aucs, delongcov)

# NOTE: this test is UNWEIGHTED. The fast DeLong algorithm assumes i.i.d. rows and has no
# survey-weighted form, so its AUCs are sample-level and will not match the survey-weighted
# 'Test AUC' column in §9. It answers "do these two models rank differently on these rows?",
# which is a model-comparison question, not a population estimate.
# Pairwise DeLong test across all fitted models, same test set
model_names = list(fitted_models.keys())
y_test_arr = y_test.to_numpy()
proba_by_model = {name: fitted_models[name].predict_proba(X_test)[:, 1] for name in model_names}

delong_rows = []
for i in range(len(model_names)):
    for j in range(i + 1, len(model_names)):
        name_i, name_j = model_names[i], model_names[j]
        (auc_i, auc_j), p = delong_roc_test(y_test_arr, proba_by_model[name_i], proba_by_model[name_j])
        delong_rows.append({
            "Model A": name_i, "Model B": name_j,
            "AUC A": round(auc_i, 4), "AUC B": round(auc_j, 4),
            "DeLong p-value": round(p, 4),
            "Significant (p<0.05)": p < 0.05,
        })

delong_df = pd.DataFrame(delong_rows)

# D20: 10 pairwise tests at alpha=0.05 inflates the family-wise false-positive rate to
# ~40% (1-(1-0.05)^10). Holm step-down correction controls this across all pairs while
# reporting the full table (this is supplementary/descriptive, not a single pre-specified
# comparison), rather than raw per-pair p<0.05 flags.
from statsmodels.stats.multitest import multipletests
_holm_reject, _holm_pvals, _, _ = multipletests(delong_df["DeLong p-value"], alpha=0.05, method='holm')
delong_df["Holm-adjusted p"] = _holm_pvals.round(4)
delong_df["Significant (Holm)"] = _holm_reject

print(f"Pairwise DeLong test -- Test AUC significance ({best_name} selected by OOF AUPRC)")
print(delong_df.to_string(index=False))

# ── Interpretation: is "best" actually distinguishable from the runner-up? ──
runner_up  = eval_df.index[1]
_pair_mask = (
    ((delong_df["Model A"] == best_name) & (delong_df["Model B"] == runner_up)) |
    ((delong_df["Model A"] == runner_up) & (delong_df["Model B"] == best_name))
)
_top_pair = delong_df[_pair_mask].iloc[0]
print(f"\nTop-2 by OOF AUPRC: {best_name} vs {runner_up} -- DeLong p = {_top_pair['DeLong p-value']:.4f} "
      f"({'significant' if _top_pair['Significant (p<0.05)'] else 'NOT significant'} at alpha=0.05).")
if not _top_pair["Significant (p<0.05)"]:
    print(f"'{best_name}' is reported as the selected model by nominal OOF AUPRC ranking only -- "
          f"it is not shown to be statistically superior to '{runner_up}'. See §9.7 below for how "
          f"this should be worded in the manuscript.")
```

<h3 style="color:#87CEEB";> 9.5 Sensitivity — Reverse-Causation Predictors </h3>

```python
# ── §9.5 Sensitivity: reverse-causation predictors (O_pain, MO4_00, OR1) ─────────
# O_pain (dental pain), MO4_00 (dental visit in the past year), and OR1 (self-rated oral
# health) are plausibly CONSEQUENCES of tooth loss rather than antecedents -- the same
# temporality objection that keeps
# brush_freq out of the confirmatory model in §11.3, and the reason O_pain was dropped from
# §11's causal exposure set.
#
# They are RETAINED in the ML arm because that arm is PREDICTIVE, not etiological: a
# downstream marker is a legitimate predictor as long as it is observable at the moment the
# model would be applied. This cell makes that defensible empirically rather than by
# assertion, refitting the SELECTED model on a feature set with both removed. Optuna
# hyperparameters are held FIXED at their §8 values -- only the feature set changes -- so any
# gap is attributable to the two variables and not to a different tuning outcome.
#
# Read it as: if discrimination barely moves, the reverse-causation objection has no
# quantitative bite. If it collapses, the model is largely restating its own outcome.
def refit_dropping(drop_cols, label):
    """Refit the SELECTED model with `drop_cols` removed, hyperparameters FIXED at §8 values.

    Only the feature set changes, so any gap against the full model is attributable to the
    dropped columns and not to a different tuning outcome. Reused by §9.6.
    """
    _feats = [c for c in feature_cols    if c not in drop_cols]
    _ordc  = [c for c in ordinal_cols    if c not in drop_cols]
    _conc  = [c for c in continuous_cols if c not in drop_cols]
    # make_imputer() names every column in feature_cols, so it cannot be reused on a
    # narrower frame -- build a matching ColumnTransformer over the reduced column lists.
    _est = build_estimator(best_name, results[best_name]['best_params'])
    _est.set_params(imputer=ColumnTransformer([
        ('mode',   SimpleImputer(strategy='most_frequent', keep_empty_features=True), _ordc),
        ('median', SimpleImputer(strategy='median',        keep_empty_features=True), _conc),
    ]))
    _Xtr, _Xte = X_train[_feats], X_test[_feats]
    _oof = cross_val_predict(_est, _Xtr, y_train, cv=cv, method='predict_proba')[:, 1]
    _est.fit(_Xtr, y_train)
    _p = _est.predict_proba(_Xte)[:, 1]
    # Column names state the weighting explicitly: OOF is the UNWEIGHTED selection metric,
    # the Test columns are SURVEY-WEIGHTED population estimates. The two are different kinds
    # of quantity and must not be read across a row -- see SENS_CAPTION below.
    return {'Feature set': f'{label} ({len(_feats)} features)',
            'OOF AUPRC (unweighted)': average_precision_score(y_train, _oof),
            'Test AUC (weighted)':    roc_auc_score(y_test, _p, sample_weight=w_test_arr),
            'Test AUPRC (weighted)':  average_precision_score(y_test, _p, sample_weight=w_test_arr)}


_p_full  = best_model.predict_proba(X_test)[:, 1]
ROW_FULL = {'Feature set': f'Full model ({len(feature_cols)} features)',
            'OOF AUPRC (unweighted)': eval_df.loc[best_name, 'OOF AUPRC'],
            'Test AUC (weighted)':    roc_auc_score(y_test, _p_full, sample_weight=w_test_arr),
            'Test AUPRC (weighted)':  average_precision_score(y_test, _p_full,
                                                             sample_weight=w_test_arr)}

# Printed under BOTH sensitivity tables (§9.5 and §9.6) so neither can be read as if the
# OOF and Test columns were the same kind of number.
SENS_CAPTION = (
    "CAPTION: 'OOF AUPRC (unweighted)' is the SELECTION metric -- unweighted, on held-out"
    "\nfolds of the 2020-2023 development sample. 'Test AUC/AUPRC (weighted)' are"
    "\nsurvey-weighted POPULATION estimates on the 2024 hold-out. These are different kinds"
    "\nof quantity: compare DOWN a column (full vs reduced), never ACROSS a row. Because"
    "\nselection is unweighted while reporting is weighted, the selected model need not have"
    "\nthe best weighted test score."
)

rc_sensitivity_df = pd.DataFrame(
    [ROW_FULL, refit_dropping(['O_pain', 'MO4_00', 'OR1'], 'Drop O_pain + MO4_00 + OR1')]
).set_index('Feature set')
rc_sensitivity_df.loc['Difference'] = rc_sensitivity_df.iloc[1] - rc_sensitivity_df.iloc[0]

print(f"§9.5 Reverse-causation sensitivity — {best_name}, hyperparameters fixed at §8 values")
print(rc_sensitivity_df.round(4).to_string())

print(SENS_CAPTION)

_d_auprc = rc_sensitivity_df.loc['Difference', 'Test AUPRC (weighted)']
_d_auc   = rc_sensitivity_df.loc['Difference', 'Test AUC (weighted)']
print(f"\nRemoving both variables changes Test AUPRC by {_d_auprc:+.4f} and Test AUC by {_d_auc:+.4f}.")
print("A small change supports keeping them in the predictive arm: discrimination does not"
      "\nrest on variables that post-date the outcome. Report this next to the main result.")
```

```python
# ── §9.6 Leave-region-out sensitivity ────────────────────────────────────────────
# region is 16 one-hot columns encoding administrative division of RESIDENCE. It is not a
# causal exposure -- no one claims a province causes tooth loss -- but it proxies things the
# model does not otherwise measure: dentist supply density, municipal water fluoridation,
# residual SES, and the urban/rural gradient beyond the binary town_t. Crude tooth-loss
# prevalence spans roughly 2.3x across the 17 divisions, so the signal is real.
#
# Two reasons to TEST it rather than trust it:
#   1. §6.1 shows region has the LARGEST distribution shift of any variable, both across the
#      COVID wave and between development and evaluation. KNHANES rotates its sample across
#      districts, so part of what region encodes is SURVEY DESIGN rather than epidemiology --
#      a generalisation risk for a predictor.
#   2. region records CURRENT residence while tooth loss accumulates over decades, so it is
#      a weak proxy for lifetime exposure.
#
# Depends on refit_dropping() and ROW_FULL from §9.5.
_region_vars = [c for c in feature_cols if c.startswith('region_')]
print(f"Dropping {len(_region_vars)} region dummies: {_region_vars}")

region_sensitivity_df = pd.DataFrame(
    [ROW_FULL, refit_dropping(_region_vars, 'Drop all region dummies')]
).set_index('Feature set')
region_sensitivity_df.loc['Difference'] = (region_sensitivity_df.iloc[1]
                                           - region_sensitivity_df.iloc[0])

print(f"\n§9.6 Leave-region-out sensitivity — {best_name}, hyperparameters fixed at §8 values")
print(region_sensitivity_df.round(4).to_string())

print(SENS_CAPTION)

_dr_auprc = region_sensitivity_df.loc['Difference', 'Test AUPRC (weighted)']
_dr_auc   = region_sensitivity_df.loc['Difference', 'Test AUC (weighted)']
print(f"\nRemoving region changes Test AUPRC by {_dr_auprc:+.4f} and Test AUC by {_dr_auc:+.4f}.")
if abs(_dr_auprc) < 0.005:
    print("NEGLIGIBLE: region is largely redundant with the SES / urbanicity variables already"
          "\nin the model, so the reported performance does not lean on geography.")
else:
    print("MATERIAL: geography carries predictive signal the other variables do not capture."
          "\nReport this as a finding about inequality in dental-care access, NOT as a"
          "\nper-province risk factor -- and note that the province code -> name mapping is"
          "\ndeliberately unverified in this notebook, so individual dummies stay unlabelled.")
```

<h3 style="color:yellow";> 9.12 SHAP-Informed Confirmatory Survey-Weighted Logistic Regression </h3>

```python
# ── SHAP-Informed Confirmatory Survey-Weighted Logistic Regression ──────────
# Bridges the ML (SHAP) and inferential (svyglm) arms: takes the top mean(|SHAP|)
# predictors from the tuned best_model and re-fits them as a design-weighted
# logistic regression (KNHANES strata/PSU/weights via R's survey::svyglm),
# giving aOR + 95% CI for exactly the variables the ML model relied on most.
# Independent of and complementary to the fully pre-specified model in §11.3.
#
# AGE IS DELIBERATELY LEFT LINEAR HERE, unlike §11 / §11.3. This cell exists to re-fit
# EXACTLY the variables the ML model relied on, and the ML feature matrix carries age as a
# single linear column -- spline-expanding it would break that correspondence and stop this
# being a faithful bridge between the two arms.
TOP_N_SHAP = 10

mean_abs_shap = np.abs(shap_values).mean(axis=0)
shap_rank = pd.Series(mean_abs_shap, index=feature_cols).sort_values(ascending=False)
top_shap_vars = shap_rank.head(TOP_N_SHAP).index.tolist()
print(f"Top {TOP_N_SHAP} SHAP-ranked variables ({best_name}): {top_shap_vars}")

# One-hot dummies (tins_/region_/occp_/house_/live_t_) aren't raw df_subset columns —
# resolve them back to a binary indicator on the parent categorical if one is selected.
_dummy_prefixes = ['tins_', 'region_', 'occp_', 'house_', 'live_t_',
                    'marri_2_', 'genertn_', 'fam_rela_']

def _resolve_column(name):
    if name in df_subset.columns:
        return df_subset[name]
    for prefix in _dummy_prefixes:
        if name.startswith(prefix):
            parent = prefix.rstrip('_')
            if name[len(prefix):] == 'nan':
                # Explicit "unknown" category introduced by §6's dummy_na=True. float('nan')
                # does NOT raise, so without this branch the line below would silently build
                # an all-zero column (x == nan is never True) and hand svyglm a constant
                # predictor. Resolve it as the parent's missingness indicator instead. It is
                # never itself NaN, so these respondents survive the complete-case .dropna()
                # below -- consistent with how the ML arm now treats them.
                return df_subset[parent].isna().astype(float)
            level = float(name[len(prefix):])
            # Preserve NaN. A bare .astype(float) turned a MISSING parent value into 0.0
            # ("not in this category"), so respondents with an unknown occupation -- 8.8%
            # of the cohort -- silently survived the complete-case .dropna() below as
            # pseudo-reference rows, inconsistently with §11, which drops them.
            return ((df_subset[parent] == level).astype(float)
                    .where(df_subset[parent].notna()))
    raise KeyError(f"Cannot resolve SHAP feature '{name}' to a df_subset column")

reg_shap = pd.DataFrame({v: _resolve_column(v) for v in top_shap_vars})
reg_shap['O_TC_'] = df_subset['O_TC_']

# Binary recodes for interpretable OR direction — same convention as §11.
# BD1 is the ODD ONE OUT: 1 = NEVER drank, 2 = Yes, so it maps {1: 0, 2: 1} while
# town_t (1 = urban) maps {1: 1, 2: 0}. marri_1 is no longer a feature (it is an exact
# function of marri_2), so it can never appear in top_shap_vars.
if 'BD1' in top_shap_vars:
    reg_shap['BD1'] = reg_shap['BD1'].map({1: 0, 2: 1})
if 'town_t' in top_shap_vars:
    reg_shap['town_t'] = reg_shap['town_t'].map({1: 1, 2: 0})
if 'BP1' in top_shap_vars:
    reg_shap['BP1'] = 5 - reg_shap['BP1']
if 'sex' in top_shap_vars:
    reg_shap['sex'] = (reg_shap['sex'] == 2).astype(float)  # 1 = female

design_cols = ['wt_itvex', 'year', 'kstrata', 'psu']
for c in design_cols:
    reg_shap[c] = df_subset[c]

m_shap = reg_shap.replace([np.inf, -np.inf], np.nan).dropna()

m_shap['svy_wt'] = m_shap['wt_itvex'] / m_shap['year'].nunique()
m_shap['svy_strata'] = pd.factorize(
    m_shap['year'].astype(int).astype(str) + '_' + m_shap['kstrata'].astype(int).astype(str)
)[0] + 1
m_shap['svy_psu'] = pd.factorize(
    m_shap['year'].astype(int).astype(str) + '_'
    + m_shap['kstrata'].astype(int).astype(str) + '_'
    + m_shap['psu'].astype(str)
)[0] + 1

print(f"Complete-case n = {len(m_shap)}  |  tooth-loss prevalence = {m_shap['O_TC_'].mean():.1%}")

import rpy2.robjects as ro
from rpy2.robjects import pandas2ri, numpy2ri
from rpy2.robjects.conversion import localconverter
from rpy2.robjects.packages import importr

survey_pkg = importr('survey')

r_cols_shap = ['O_TC_', 'svy_wt', 'svy_strata', 'svy_psu'] + top_shap_vars
with localconverter(ro.default_converter + numpy2ri.converter + pandas2ri.converter):
    r_df_shap = pandas2ri.py2rpy(m_shap[r_cols_shap])

ro.r('options(survey.lonely.psu = "adjust")')
svy_design_shap = survey_pkg.svydesign(
    ids=ro.Formula('~svy_psu'),
    strata=ro.Formula('~svy_strata'),
    weights=ro.Formula('~svy_wt'),
    data=r_df_shap
)

formula_str_shap = 'O_TC_ ~ ' + ' + '.join(top_shap_vars)
svyfit_shap = survey_pkg.svyglm(
    ro.Formula(formula_str_shap),
    design=svy_design_shap,
    family=ro.r('quasibinomial(link="logit")')
)

coef_mat_shap = ro.r('summary')(svyfit_shap).rx2('coefficients')
ci_mat_shap = ro.r('confint')(svyfit_shap, level=0.95)
with localconverter(ro.default_converter + numpy2ri.converter + pandas2ri.converter):
    coef_df_shap = pd.DataFrame(
        pandas2ri.rpy2py(coef_mat_shap),
        index=list(ro.r('rownames')(coef_mat_shap)),
        columns=['Estimate', 'SE', 't', 'p-value']
    )
    ci_df_shap = pd.DataFrame(
        pandas2ri.rpy2py(ci_mat_shap),
        index=list(ro.r('rownames')(ci_mat_shap)),
        columns=['CI 2.5%', 'CI 97.5%']
    )

or_table_shap = pd.DataFrame({
    'aOR':      np.exp(coef_df_shap['Estimate']),
    'CI 2.5%':  np.exp(ci_df_shap['CI 2.5%']),
    'CI 97.5%': np.exp(ci_df_shap['CI 97.5%']),
    'p-value':  coef_df_shap['p-value']
})

# ── Multiplicity correction ──────────────────────────────────────────────────
# Same Holm (FWER) + Benjamini-Hochberg (FDR) treatment as §11, over all non-intercept
# terms. §11.3 is deliberately left uncorrected because its covariates are pre-specified;
# this model's are not, so it belongs with §11 rather than with §11.3.
from statsmodels.stats.multitest import multipletests

_terms_shap = [t for t in or_table_shap.index if t != '(Intercept)']
_p_raw_shap = or_table_shap.loc[_terms_shap, 'p-value'].astype(float).to_numpy()
or_table_shap['Holm p'] = np.nan
or_table_shap['BH q']   = np.nan
or_table_shap.loc[_terms_shap, 'Holm p'] = multipletests(_p_raw_shap, alpha=0.05, method='holm')[1]
or_table_shap.loc[_terms_shap, 'BH q']   = multipletests(_p_raw_shap, alpha=0.05, method='fdr_bh')[1]
or_table_shap['Survives'] = np.select(
    [or_table_shap.index == '(Intercept)',
     or_table_shap['Holm p'] < 0.05,
     or_table_shap['BH q'] < 0.05,
     or_table_shap['p-value'] < 0.05],
    ['', 'Holm+BH', 'BH only', 'raw only'],
    default='')

print(f"\n=== SHAP-informed confirmatory model — top {TOP_N_SHAP} predictors (survey::svyglm) ===")
print(f"{len(_terms_shap)} non-intercept terms | multiplicity: Holm (FWER) + Benjamini-Hochberg (FDR)")
print(or_table_shap.round(4).to_string())
print("\nCAVEAT: these predictors were SELECTED by SHAP, so the p-values are POST-SELECTION."
      "\nHolm/BH control the number of tests performed, not the fact that only the winners"
      "\nwere tested. Read this table as exploratory; §11.3 is the pre-specified model.")
```

Diagnostic plots for the best tuned model on the **test set**, built with `sklearn` + `matplotlib` (no PyCaret). The confusion matrix uses the top model from §9 (feature importance via SHAP is covered in §9, not here); §10.1 overlays the ROC curves of all tuned models.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, roc_curve, matthews_corrcoef
from sklearn.model_selection import cross_val_predict

# ── Confusion matrices — best model, 3 thresholds (Default 0.5 / Youden's J / MCC) ──
# Threshold selection via OOF probabilities — avoids optimising on the test set
oof_proba = cross_val_predict(best_model, X_train, y_train,
                              cv=cv, method='predict_proba')[:, 1]
fpr_oof, tpr_oof, thresh_oof = roc_curve(y_train, oof_proba)

# Youden's J threshold
j_scores  = tpr_oof + (1 - fpr_oof) - 1
best_idx  = j_scores.argmax()
THRESHOLD = float(thresh_oof[best_idx])
print(f"Youden's J threshold (OOF train): {THRESHOLD:.4f}")
print(f"  OOF Sensitivity: {tpr_oof[best_idx]:.4f}")
print(f"  OOF Specificity: {1 - fpr_oof[best_idx]:.4f}")

def mcc_curve(y_true, proba, thresholds):
    """MCC at every threshold, vectorised.

    Exactly equivalent to [matthews_corrcoef(y_true, (proba >= t)) for t in thresholds],
    but O(n log n) instead of O(n * n_thresholds) -- the loop rebuilt an n-length prediction
    array for each of ~n distinct thresholds, which dominated the runtime of this cell and
    of the 3-model grid below.

    Sort the scores descending once; as the threshold sweeps down, rows flip 0 -> 1 in that
    order, so the confusion matrix at the k-th cut-point is a running total. This is the same
    trick sklearn's own roc_curve uses to build its TPS/FPS arrays.
    """
    y = np.asarray(y_true, dtype=np.int64)
    p = np.asarray(proba, dtype=float)
    order    = np.argsort(-p, kind='mergesort')
    y_sorted = y[order]
    p_sorted = p[order]
    P = int(y.sum())
    N = y.size - P
    # k = how many rows are predicted positive at each threshold, i.e. count of p >= t
    k      = np.searchsorted(-p_sorted, -np.asarray(thresholds, dtype=float), side='right')
    cum_tp = np.concatenate(([0], np.cumsum(y_sorted)))
    TP = cum_tp[k].astype(float)
    FP = k - TP
    FN = P - TP
    TN = N - FP
    denom = np.sqrt((TP + FP) * (TP + FN) * (TN + FP) * (TN + FN))
    with np.errstate(invalid='ignore', divide='ignore'):
        mcc = (TP * TN - FP * FN) / denom
    return np.where(denom == 0, 0.0, mcc)   # sklearn returns 0 for a degenerate split


mcc_scores    = mcc_curve(y_train, oof_proba, thresh_oof)
best_mcc_idx  = int(np.argmax(mcc_scores))
MCC_THRESHOLD = float(thresh_oof[best_mcc_idx])
best_mcc_oof  = float(mcc_scores[best_mcc_idx])
print(f"MCC threshold (OOF train):  {MCC_THRESHOLD:.4f}")
print(f"Best OOF MCC score:          {best_mcc_oof:.4f}")

# Apply each frozen threshold once to the held-out test set
proba_best = best_model.predict_proba(X_test)[:, 1]
thresholds = [0.5, THRESHOLD, MCC_THRESHOLD]
col_labels = ["Default (0.5)", "Youden's J", 'MCC']

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, thresh, label in zip(axes, thresholds, col_labels):
    preds = (proba_best >= thresh).astype(int)
    cm    = confusion_matrix(y_test, preds)
    ConfusionMatrixDisplay(
        cm, display_labels=['≥20 teeth', '<20 teeth']
    ).plot(ax=ax, cmap='Blues', colorbar=False)
    t_str = '' if thresh == 0.5 else f' (t={thresh:.3f})'
    ax.set_title(f'{label}{t_str}')
    if ax is not axes[0]:
        ax.set_ylabel('')

plt.suptitle(f'Confusion matrix — {best_name} (test set)', fontsize=13)
plt.tight_layout()
plt.show()
```

<h3 style="color:yellow";> 10.3 Post-hoc Recalibration — Logistic (Platt) Scaling </h3>

`class_weight="balanced"` / `scale_pos_weight` shift predicted probabilities away from true prevalence — the same direction SMOTE would, which §11 explicitly avoided SMOTE for. Recalibration is fit on out-of-fold **train** probabilities (`oof_proba_by_model`, §7) only; the test set is untouched until the final comparison below. The recalibration model's own coefficients (slope, intercept) double as the calibration diagnostics.


```python
# ── Post-hoc recalibration (Platt / logistic scaling) ──
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import brier_score_loss

def _logit(p, eps=1e-6):
    p = np.clip(p, eps, 1 - eps)
    return np.log(p / (1 - p))

recal_rows = []
recal_proba_by_model = {}
for name, model in fitted_models.items():
    proba_raw = model.predict_proba(X_test)[:, 1]

    # C=np.inf: Platt scaling must be UNPENALISED. LogisticRegression() defaults to
    # L2 with C=1, which shrinks the fitted slope toward 0 -- and the slope/intercept are
    # themselves reported below as the calibration diagnostic, so a shrunk slope would
    # misstate the very quantity this cell exists to measure.
    # C=np.inf replaces the deprecated penalty=None (removed in scikit-learn 1.10);
    # coefficients are bit-identical. sklearn 1.8 emits a cosmetic UserWarning here.
    recalibrator = LogisticRegression(C=np.inf).fit(
        _logit(oof_proba_by_model[name]).reshape(-1, 1), y_train)
    slope, intercept = recalibrator.coef_[0, 0], recalibrator.intercept_[0]
    proba_recal = recalibrator.predict_proba(_logit(proba_raw).reshape(-1, 1))[:, 1]
    recal_proba_by_model[name] = proba_recal

    recal_rows.append({
        "Model": name,
        "Slope": slope,
        "Intercept": intercept,
        # Test Brier scores are survey-weighted; the recalibrator itself is FITTED on
        # unweighted training OOF probabilities, since fitting stays unweighted throughout.
        "Brier (raw)": brier_score_loss(y_test, proba_raw, sample_weight=w_test_arr),
        "Brier (recalibrated)": brier_score_loss(y_test, proba_recal, sample_weight=w_test_arr),
    })

recal_df = pd.DataFrame(recal_rows).set_index("Model")
print(recal_df.round(4).to_string())
```

<h3 style="color:yellow";> 10.4 Decision Curve Analysis — Clinical Utility (Recalibrated Probabilities) </h3>

AUC and AUPRC measure how well a model *ranks* respondents; neither says whether acting on a predicted risk is worth doing. Decision curve analysis (Vickers & Elkin, 2006) reports **net benefit** across a range of referral thresholds, against the two trivial defaults — refer everyone, refer no one.

**This section deliberately sits after §10.3.** Net benefit is evaluated *at* a probability threshold, so it is only meaningful if a predicted risk of 0.20 really corresponds to a 20% chance. Every tuned model here is fitted with `class_weight="balanced"` / `scale_pos_weight`, which §10.3 shows pushes predicted probabilities away from true prevalence — the same distortion that ruled out SMOTE. DCA is therefore computed on `recal_proba_by_model` (Platt-scaled on out-of-fold **training** probabilities), not on raw model output. Running it on the raw probabilities, as an earlier version did, gave net-benefit values that were not interpretable at any threshold.

```python
# ── Decision Curve Analysis — clinical utility of using each model to guide referral decisions ──
# AUC/AUPRC measure ranking quality; DCA measures whether acting on a model's predicted risk at a
# given threshold does more good than the "refer everyone" / "refer no one" defaults (Vickers &
# Elkin, 2006).
#
# USES THE RECALIBRATED PROBABILITIES from §10.3, not the raw ones. Net benefit is evaluated
# AT a probability threshold, so it is only interpretable if the predicted probability means
# what it says. Every model here is fitted with class_weight="balanced" / scale_pos_weight,
# which §10.3 shows deliberately shifts predicted probabilities away from true prevalence --
# exactly the distortion that made SMOTE unacceptable. Running DCA on those raw probabilities
# produced net-benefit numbers that were not interpretable at any threshold. This cell now
# sits AFTER §10.3 so recal_proba_by_model is available.
from dcurves import dca, plot_graphs

# NOTE: dcurves accepts no survey weights, so net benefit here is a SAMPLE-level quantity,
# unlike the survey-weighted metrics in §9. Report it as such.
dca_thresholds = np.arange(0.01, 0.51, 0.01)

dca_data = pd.DataFrame({"O_TC_": y_test_arr})
for name in model_names:
    dca_data[name] = recal_proba_by_model[name]

dca_result = dca(
    data=dca_data,
    outcome="O_TC_",
    modelnames=model_names,
    thresholds=dca_thresholds,
)

y_max = dca_result["net_benefit"].max() + 0.05
plt.figure(figsize=(8, 6))
plt.title(f"Decision Curve Analysis — Test Set, recalibrated probabilities (n={len(y_test_arr)})")
plot_graphs(
    plot_df=dca_result,
    graph_type="net_benefit",
    y_limits=(-0.05, y_max),
)
```

```python
# ── DCA results as text (net benefit by threshold; recalibrated probabilities) ──
dca_table = (
    dca_result
    .pivot(index="threshold", columns="model", values="net_benefit")
    .round(4)
)
dca_table = dca_table[model_names + ["all", "none"]]
pd.set_option("display.max_rows", None)
print(dca_table.to_string())
```

```python
# ── Age spline: basis, knot selection, and helpers (used by §11 and §11.3) ──────────
# Age enters the survey-weighted models below as a RESTRICTED CUBIC SPLINE rather than a
# single linear term. Over a 30-year span the log-odds of tooth loss are not plausibly
# linear in age, and age is the strongest predictor in every model here, so a linear-only
# term was the most likely misspecification in the confirmatory analysis.
#
# The ML arm is deliberately NOT changed. The tree models capture nonlinearity natively,
# and spline-expanding age for its Logistic Regression alone would mean the five candidate
# models no longer see the same feature matrix, breaking the head-to-head comparison in §9.
#
# Basis: Harrell's restricted cubic spline (Regression Modeling Strategies §2.4.4). With k
# knots it contributes k-1 terms -- a linear term, which is numerically identical to raw age
# and is therefore the 'age' column already in the model, plus k-2 nonlinear terms.
# "Restricted" means the fit is forced linear beyond the outer knots, so the tails cannot
# swing wildly where the data thins out.
import statsmodels.api as sm

# Harrell's recommended knot QUANTILES by knot count. Placement depends only on the age
# distribution, never on the outcome.
HARRELL_KNOT_QUANTILES = {
    3: [0.10, 0.50, 0.90],
    4: [0.05, 0.35, 0.65, 0.95],
    5: [0.05, 0.275, 0.50, 0.725, 0.95],
}


def rcs_basis(x, knots):
    """Harrell restricted cubic spline basis.

    Returns an (n, k-1) array: column 0 is x itself (the linear term), columns 1..k-2 are
    the nonlinear terms.
    """
    x = np.asarray(x, dtype=float)
    t = np.asarray(knots, dtype=float)
    k = len(t)
    if k < 3:
        raise ValueError('a restricted cubic spline needs at least 3 knots')
    cube  = lambda u: np.where(u > 0, u ** 3, 0.0)
    denom = (t[-1] - t[0]) ** 2
    cols  = [x]
    for j in range(k - 2):
        cols.append((cube(x - t[j])
                     - cube(x - t[k - 2]) * (t[k - 1] - t[j]) / (t[k - 1] - t[k - 2])
                     + cube(x - t[k - 1]) * (t[k - 2] - t[j]) / (t[k - 1] - t[k - 2])) / denom)
    return np.column_stack(cols)


def add_age_spline(frame, model_vars, knots):
    """Append the NONLINEAR age-spline columns to `frame` and to `model_vars`.

    'age' is already carried as the linear term (rcs_basis column 0 == age), so only the
    k-2 nonlinear columns are added. Returns (frame, new_model_vars, nl_col_names).
    """
    basis   = rcs_basis(frame['age'].to_numpy(), knots)
    nl_cols = [f'age_rcs{j + 2}' for j in range(basis.shape[1] - 1)]
    for j, c in enumerate(nl_cols):
        frame[c] = basis[:, j + 1]
    return frame, list(model_vars) + nl_cols, nl_cols


def wald_nonlinearity(fit, nl_cols, label):
    """Design-based Wald test that the nonlinear age terms are JOINTLY zero.

    This is the reportable answer to "is the age effect actually nonlinear?" -- it uses the
    survey design (Taylor linearization), not a naive likelihood-ratio test, which is not
    valid under weighting. Guarded so a survey-package quirk cannot kill the whole cell.
    """
    try:
        rt = survey_pkg.regTermTest(fit, ro.Formula('~' + ' + '.join(nl_cols)))
        p  = float(np.asarray(rt.rx2('p')).ravel()[0])
        verdict = 'NONLINEAR (linearity rejected)' if p < 0.05 else 'no evidence against linearity'
        print(f"{label} — design-based Wald test of age nonlinearity: p = {p:.4g}  ->  {verdict}")
        return p
    except Exception as exc:
        print(f"{label} — nonlinearity test unavailable ({type(exc).__name__}: {exc})")
        return np.nan


# ── Knot-count selection by AIC ────────────────────────────────────────────────────
# Marginal (covariate-free, unweighted) logistic fit of the outcome on the age basis alone.
# Knot COUNT is a modelling choice fixed before any inference; leaving covariates out stops
# the count being driven by the adjustment set, and quantile placement already makes the
# knot POSITIONS outcome-independent. Selection uses all five waves, matching the survey
# models below, which have no hold-out.
def _knot_positions(vals, quantiles):
    """Knot positions for one candidate knot count.

    Harrell places knots at quantiles of the predictor. A narrow or heavily top-coded age
    range can make two quantiles land on the same age, which gives a singular basis --
    KNHANES top-codes age at 80, and the >=65 cohort spans only ~16 distinct values, so this
    is a live risk here rather than a theoretical one. Fall back to equally-spaced knots
    across the observed range in that case, and report which rule was used.
    """
    kn = np.quantile(vals, quantiles)
    if len(np.unique(kn)) == len(quantiles):
        return kn, 'quantile'
    kn = np.linspace(vals.min(), vals.max(), len(quantiles))
    if len(np.unique(kn)) == len(quantiles):
        return kn, 'equally-spaced (quantiles tied)'
    return None, 'unusable'


_age_sel  = df_subset[['age', 'O_TC_']].dropna()
_age_vals = _age_sel['age'].to_numpy()
_age_y    = _age_sel['O_TC_'].to_numpy()
_aic_rows, _knot_sets = [], {}
for _k, _q in HARRELL_KNOT_QUANTILES.items():
    _kn, _how = _knot_positions(_age_vals, _q)
    if _kn is None:
        _aic_rows.append({'knots': _k, 'placement': _how, 'positions': '—', 'AIC': np.nan})
        continue
    _knot_sets[_k] = _kn
    _Xk  = sm.add_constant(rcs_basis(_age_vals, _kn), has_constant='add')
    _fit = sm.Logit(_age_y, _Xk).fit(disp=0)
    _aic_rows.append({'knots': _k, 'placement': _how,
                      'positions': ', '.join(f'{v:.1f}' for v in _kn),
                      'AIC': _fit.aic})

_aic_df = pd.DataFrame(_aic_rows).set_index('knots')
assert _aic_df['AIC'].notna().any(), (
    'no usable age-spline basis: age has fewer than 3 distinct values in this cohort')
_best_k   = int(_aic_df['AIC'].idxmin())
AGE_KNOTS = _knot_sets[_best_k]

print(f"Age-spline knot selection — marginal logistic fit, n = {len(_age_sel):,}, "
      f"age {_age_vals.min():.0f}-{_age_vals.max():.0f}")
print(_aic_df.round(2).to_string())
print(f"\nSelected {_best_k} knots at ages {', '.join(f'{v:.1f}' for v in AGE_KNOTS)} (lowest AIC).")
print(f"Contributes the existing linear 'age' term + {_best_k - 2} nonlinear term(s): "
      f"{', '.join(f'age_rcs{j + 2}' for j in range(_best_k - 2))}")
```

<h2 style="color:#FFA500;"> 11. Inferential Analysis — Multivariable Logistic Regression </h2>

The sections above answer *"what predicts tooth loss?"* (prediction). The research question is different — *"which clinical, behavioral, and socioeconomic factors affect tooth loss **independently of age and sex**?"* That is an **explanatory** question, answered with logistic regression reporting **adjusted odds ratios (aOR)**.

<h4 style="color:#7FDBFF";> Note — quasi-complete separation in the table above </h4>

Several nominal categories above (e.g. `fam_rela` levels with very few respondents in this cohort) have too few observations for a stable coefficient estimate. An `aOR` at or near 0, or implausibly large, is a **separation artifact**, not a robust finding — even when it is flagged `Holm+BH` in the `Survives` column (surviving multiplicity correction does not fix a degenerate estimate). This is the same failure mode the `_degenerate` check under §11.3.1 below guards against for the pre-specified model; §11's own table has no equivalent guard. Treat any `Survives = Holm+BH` row with `aOR ≤ 0.05` or `aOR ≥ 20` as unreliable pending a category-size check, not as the strongest finding in this table.

```python
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Build modelling frame from the cleaned df_subset.
# sex is now retained in df_subset after eGFR derivation, so no need to pull from original df.
reg = df_subset.copy()
reg['sex_female'] = (reg['sex'] == 2).astype(float)
# 'diabetes' column already present — combined criterion (DE1_dg == 1 OR HbA1c >= 6.5%) created in §4.3

# Binary recodes: KNHANES codes 2=No → recode to 0 for intuitive OR direction (1=yes vs 0=no)
# BD1 is LIFETIME DRINKING EXPERIENCE coded 1 = NEVER, 2 = Yes, so the drinker
# indicator maps {1: 0, 2: 1}. town_t (1 = urban) maps the other way, {1: 1, 2: 0}.
# Folding both into one loop inverted BD1 and labelled never-drinkers as drinkers.
reg['BD1']    = reg['BD1'].map({1: 0, 2: 1})
reg['town_t'] = reg['town_t'].map({1: 1, 2: 0})
# BP1 (perceived stress): original 1=very stressed → 4=not at all; reverse so higher = more stressed
reg['BP1'] = 5 - reg['BP1']

exposures = [
    # core (existing)
    'eGFR', 'diabetes', 'ho_incm', 'edu',
    # additional SES
    # marri_1 dropped: with marri_2 carrying an explicit never-married level (0), marri_1 is
    # an exact function of it (marri_1 == 2 <=> marri_2 == 0). Keeping both made marri_1
    # zero-variance in the complete-case frame -- the guard below removed it silently -- and
    # dropped every never-married respondent from the model. The contrast is now marri_2_0.
    # region ADDED as dummies (see the nominal block below).
    'occp', 'marri_2', 'genertn', 'fam_rela', 'town_t', 'region',
    # lifestyle
    'sm_presnt', 'BD1', 'pa_aerobic', 'BP1',
    # clinical
    'HE_BMI', 'HE_sbp', 'HE_dbp', 'HE_Uacid', 'HE_chol', 'HE_HDL_st2',
    # urinalysis
    'HE_Upro', 'HE_Ucrea',
    # dental
    # O_pain, MO4_00, and OR1 REMOVED: dental pain, dental visit in the past year, and
    #   self-rated oral health are plausibly CONSEQUENCES of tooth loss rather than
    #   antecedents -- the same temporality objection that keeps brush_freq out of §11.3.
    #   §11 is framed causally (aORs, confounding language), so a reverse-caused term does
    #   not belong here. All three are RETAINED in the ML arm, which is predictive rather
    #   than etiological; §9.5 quantifies how much discrimination rests on them.
    # lipids
    'HE_TG',
    # dietary
    'N_CA', 'N_NA', 'N_PROT', 'N_TDF', 'LS_VEG1', 'LS_FRUIT', 'N_SUGAR',
]
# region     NOW INCLUDED as 16 dummies (reference = code 1). The old exclusion note said
#            "no linear ordering, not valid as continuous" -- true, but this cell already
#            dummy-encodes occp / marri_2 / genertn / fam_rela, so the same machinery
#            applies. Design df (nPSU - nStrata) is ~700 here, ample for 16 extra terms.
# HE_wc     excluded: collinear with HE_BMI (VIF 5.81); HE_BMI retained
# L_BR_FQ   excluded: crude tooth-loss rate by category is non-monotonic (not computed/verified in this notebook) — linear coefficient uninterpretable

covariates = ['age', 'sex_female']
model_vars = exposures + covariates

# complete-case; NO SMOTE (would bias the odds ratios)
# Survey design columns pulled alongside model vars for weighted inference
design_cols = ['wt_itvex', 'year', 'kstrata', 'psu']
m = (reg[['O_TC_'] + design_cols + model_vars]
     .replace([np.inf, -np.inf], np.nan).dropna())

# KNHANES integrated weight for pooled waves: annual weight / #years actually present
m['svy_wt'] = m['wt_itvex'] / m['year'].nunique()
# strata/PSU codes repeat across waves -> make globally unique per (year, stratum/PSU)
# survey internals coerce strata/PSU with as.numeric() -> string IDs become NA
# and blow up the variance estimate (Inf SE, p=1.0). Use integer codes instead.
m['svy_strata'] = pd.factorize(
    m['year'].astype(int).astype(str) + '_' + m['kstrata'].astype(int).astype(str)
)[0] + 1
m['svy_psu'] = pd.factorize(
    m['year'].astype(int).astype(str) + '_'
    + m['kstrata'].astype(int).astype(str) + '_'
    + m['psu'].astype(str)
)[0] + 1

print(f"Complete-case n = {len(m)}  |  tooth-loss prevalence = {m['O_TC_'].mean():.1%}")

# Nominal vars -> dummy-encode with an EXPLICIT reference level, replacing each single
# linear term so every category gets its own OR. Named explicitly rather than via
# drop_first, which silently takes whatever the lowest OBSERVED code happens to be --
# that reference can shift between the two cohort notebooks and change what every
# coefficient in the block means. These match the ML arm's `nominal_ref` in §6.
#   occp 1 = Managers/Professionals | marri_2 1 = married-cohabiting (0 = never married)
#   genertn 1 | fam_rela 1 = household head | region 1 = first province code
nominal_ref_inf = {'occp': 1, 'marri_2': 1, 'genertn': 1, 'fam_rela': 1, 'region': 1}
for _nom, _ref in nominal_ref_inf.items():
    if _nom in model_vars:
        _nom_d = pd.get_dummies(m[_nom].round().astype(int), prefix=_nom, dtype=float)
        _ref_col = f'{_nom}_{_ref}'
        assert _ref_col in _nom_d.columns, (
            f"reference level {_ref_col} not present in the complete-case data; "
            f"available: {list(_nom_d.columns)}")
        _nom_d = _nom_d.drop(columns=[_ref_col])
        m = pd.concat([m, _nom_d], axis=1)
        model_vars = [c for c in model_vars if c != _nom] + list(_nom_d.columns)
        print(f"  {_nom}: {len(_nom_d.columns)} dummies, reference = {_ref_col}")

# Age enters as a RESTRICTED CUBIC SPLINE (basis + AIC-selected knots from the cell above).
# 'age' is already in model_vars as the spline's linear term, so only the nonlinear columns
# are appended. Its coefficient is therefore no longer "the" age effect on its own -- the
# age association is the linear term and the nonlinear terms read together.
m, model_vars, _age_nl = add_age_spline(m, model_vars, AGE_KNOTS)
print(f"  age spline: +{len(_age_nl)} nonlinear term(s) {_age_nl}"
      f"  (knots at {', '.join(f'{v:.1f}' for v in AGE_KNOTS)})")

# Drop zero-variance columns (inestimable in any GLM)
# std < 1e-8 catches near-constant float columns that nunique() misses
zero_var = [c for c in model_vars if m[c].nunique() <= 1 or m[c].std() < 1e-8]
if zero_var:
    print(f"Dropping zero-variance columns (inestimable): {zero_var}")
    model_vars = [c for c in model_vars if c not in zero_var]

# Full complex-survey logistic regression via R's survey::svyglm
# Handles strata + PSU + weights with Taylor linearization — fully design-based SEs
import rpy2.robjects as ro
from rpy2.robjects import pandas2ri, numpy2ri
from rpy2.robjects.conversion import localconverter
from rpy2.robjects.packages import importr

survey_pkg = importr('survey')

# Pass model frame + design columns to R
r_cols = ['O_TC_', 'svy_wt', 'svy_strata', 'svy_psu'] + model_vars

# Narrow context: only convert the DataFrame going into R
with localconverter(ro.default_converter + numpy2ri.converter + pandas2ri.converter):
    r_df = pandas2ri.py2rpy(m[r_cols])

# R function calls outside the converter context so their return values
# (R S3 objects) are NOT auto-converted back to Python lists
# Strata with a single PSU can't estimate variance by default; 'adjust'
# centers the unit at the overall mean (Lumley's recommended handling)
ro.r('options(survey.lonely.psu = "adjust")')
svy_design = survey_pkg.svydesign(
    ids=ro.Formula('~svy_psu'),
    strata=ro.Formula('~svy_strata'),
    weights=ro.Formula('~svy_wt'),
    data=r_df
)

# quasibinomial recommended by Lumley for survey-weighted binary outcomes
formula_str = 'O_TC_ ~ ' + ' + '.join(model_vars)
svyfit = survey_pkg.svyglm(
    ro.Formula(formula_str),
    design=svy_design,
    family=ro.r('quasibinomial(link="logit")')
)

# Narrow context: only convert the result matrices back to pandas
coef_mat = ro.r('summary')(svyfit).rx2('coefficients')
ci_mat = ro.r('confint')(svyfit, level=0.95)
with localconverter(ro.default_converter + numpy2ri.converter + pandas2ri.converter):
    coef_df = pd.DataFrame(
        pandas2ri.rpy2py(coef_mat),
        index=list(ro.r('rownames')(coef_mat)),
        columns=['Estimate', 'SE', 't', 'p-value']
    )
    ci_df = pd.DataFrame(
        pandas2ri.rpy2py(ci_mat),
        index=list(ro.r('rownames')(ci_mat)),
        columns=['CI 2.5%', 'CI 97.5%']
    )

or_table = pd.DataFrame({
    'aOR':      np.exp(coef_df['Estimate']),
    'CI 2.5%':  np.exp(ci_df['CI 2.5%']),
    'CI 97.5%': np.exp(ci_df['CI 97.5%']),
    'p-value':  coef_df['p-value']
})

# ── Multiplicity correction (§11 is the FULL EXPLORATORY model) ─────────────
# This model estimates a large, un-restricted coefficient set, so at alpha=0.05 several
# nominally significant rows are expected by chance alone. Holm controls the family-wise
# error rate; Benjamini-Hochberg controls the false discovery rate. Both are reported over
# all non-intercept terms so a reader can see which findings survive which criterion.
# §11.3 (pre-specified, a priori) is deliberately NOT corrected -- its covariates were not
# selected by any search, so its p-values are not inflated by multiplicity.
from statsmodels.stats.multitest import multipletests

_terms = [t for t in or_table.index if t != '(Intercept)']
_p_raw = or_table.loc[_terms, 'p-value'].astype(float).to_numpy()
or_table['Holm p'] = np.nan
or_table['BH q']   = np.nan
or_table.loc[_terms, 'Holm p'] = multipletests(_p_raw, alpha=0.05, method='holm')[1]
or_table.loc[_terms, 'BH q']   = multipletests(_p_raw, alpha=0.05, method='fdr_bh')[1]
or_table['Survives'] = np.select(
    [or_table.index == '(Intercept)',
     or_table['Holm p'] < 0.05,
     or_table['BH q'] < 0.05,
     or_table['p-value'] < 0.05],
    ['', 'Holm+BH', 'BH only', 'raw only'],
    default='')

print("\n=== Adjusted odds ratios (survey::svyglm, Taylor linearization) ===")
print(f"{len(_terms)} non-intercept terms | multiplicity: Holm (FWER) + Benjamini-Hochberg (FDR)")
print(or_table.round(4).to_string())
print("\n'Survives': Holm+BH = significant after family-wise correction (strongest claim);"
      "\n             BH only  = survives FDR but not FWER;"
      "\n             raw only = nominally p<0.05 but survives NEITHER correction.")
print(f"  nominal p<0.05: {(or_table.loc[_terms, 'p-value'] < 0.05).sum()}"
      f"  |  BH q<0.05: {(or_table.loc[_terms, 'BH q'] < 0.05).sum()}"
      f"  |  Holm p<0.05: {(or_table.loc[_terms, 'Holm p'] < 0.05).sum()}")

print()
wald_nonlinearity(svyfit, _age_nl, '§11 full exploratory model')
```

<h3 style="color:yellow";> 11.3 Pre-Specified Survey-Weighted Logistic Regression — a priori covariate set </h3>

The covariate set below was **specified a priori from prior evidence** — prior literature (systemic, socioeconomic/spatial, behavioral-nutritional theory domains) — **before and independently of the SHAP analysis**. No variable here was chosen from SHAP output, model importances, or any outcome-driven screen.

This breaks the circularity in the *SHAP-Informed Confirmatory* model, which selects the top mean(|SHAP|) predictors and then "confirms" them. Because the ML/SHAP arm and this LR now choose their variables by **independent criteria** (data-driven vs. theory-driven), agreement between them is **genuine triangulation**, not circular self-confirmation.

**Pre-specified covariates (18), by theory block:**

- **Systemic:** diabetes, eGFR, hypertension, uric acid (`HE_Uacid`), BMI (`HE_BMI`), triglycerides (`HE_TG`)
- **Socioeconomic / spatial:** sex, age, income (`ho_incm`), education (`edu`), urbanicity (`town_t`), occupation (`occp`)
- **Behavioral / nutrition:** smoking (`sm_presnt`), alcohol (`BD1` — note KNHANES codes `1 = never`, so the drinker indicator is `{1: 0, 2: 1}`), physical activity (`pa_aerobic`), stress (`BP1`), dietary fiber (`N_TDF`), calcium (`N_CA`)

**Brushing is excluded from the confirmatory model.** It is recorded for *yesterday* — after the outcome accumulated — and it overlaps structurally with edentulism (`O_TC == 0` ⇒ `O_TC_ == 1` by construction), so a confirmatory aOR would blend a behavioural effect with a definitional one. Brushing remains an ML feature and may enter the **exploratory** SHAP-informed regression in §9 if SHAP surfaces it, which is the appropriate place for it. Excluding it here also keeps every edentulous respondent in the sample; the earlier NaN-based handling silently dropped them via the complete-case filter, which biased the confirmatory model's outcome prevalence downward, since every edentulous respondent is a case by construction (see the retained sample size and prevalence printed by the cell below).

Design-based inference (KNHANES strata / PSU / weights via `survey::svyglm`, Taylor linearization) is identical to §11.

```python
# ── §11.3 Pre-Specified Survey-Weighted Logistic Regression (a priori) ───────
# Covariates fixed from PRIOR LITERATURE (systemic / socioeconomic-spatial / behavioral-nutritional theory domains) BEFORE and
# INDEPENDENTLY of SHAP. This cell references NO SHAP object — the variable list
# is theory-driven, so agreement with the ML/SHAP arm is genuine triangulation,
# not circular confirmation. Same survey design + svyglm machinery as §11.

# Fresh frame straight from cleaned df_subset (self-contained; does NOT depend on
# §11's mutated `reg`/`m`).
reg2 = df_subset.copy()
reg2['sex_female'] = (reg2['sex'] == 2).astype(float)
# KNHANES 2=No -> 0 for intuitive OR direction (1=yes vs 0=no).
# BD1 is the EXCEPTION: lifetime drinking experience is coded 1 = NEVER, 2 = Yes, so the
# drinker indicator maps {1: 0, 2: 1}. town_t (1 = urban) maps {1: 1, 2: 0}.
reg2['BD1']    = reg2['BD1'].map({1: 0, 2: 1})
reg2['town_t'] = reg2['town_t'].map({1: 1, 2: 0})
# BP1 (perceived stress): 1=very stressed ... 4=not at all -> reverse so higher = more stressed
reg2['BP1'] = 5 - reg2['BP1']
# 'diabetes' and 'hypertension' are already 0/1 binary flags (derived in §4)

# Pre-specified covariate set (theory-driven; NOT from SHAP) -------------------
prespec_systemic   = ['diabetes', 'eGFR', 'hypertension', 'HE_Uacid', 'HE_BMI', 'HE_TG']
prespec_ses        = ['sex_female', 'age', 'ho_incm', 'edu', 'town_t', 'occp']
prespec_behavioral = ['sm_presnt', 'BD1', 'pa_aerobic', 'BP1', 'N_TDF', 'N_CA']
prespec_vars = prespec_systemic + prespec_ses + prespec_behavioral
# brush_freq DELIBERATELY EXCLUDED from the confirmatory model:
#   (a) TEMPORALITY -- brushing is recorded for "yesterday", i.e. after the outcome
#       accumulated over decades, so it cannot be read as an antecedent risk factor;
#   (b) STRUCTURAL OVERLAP -- O_TC == 0 => O_TC_ == 1 by construction, and edentulous
#       respondents partly populate the non-brushing group, so part of any brushing
#       coefficient is definitional rather than behavioural (quantified in §4.3).
# Brushing stays in the ML arm (BM1_0 + brush_freq in feature_cols) and may enter the
# EXPLORATORY SHAP-informed regression in §9 if SHAP surfaces it -- the right venue.
# Excluding it here also retains all edentulous respondents: the earlier NaN-based
# handling dropped them via the complete-case filter, and because every one of them is
# a case by construction, that was selection on the outcome inside the model designated
# as confirmatory. (Retained sample size and prevalence are printed below.)

design_cols = ['wt_itvex', 'year', 'kstrata', 'psu']
m2 = (reg2[['O_TC_'] + design_cols + prespec_vars]
      .replace([np.inf, -np.inf], np.nan).dropna())

# KNHANES integrated weight for pooled waves; globally-unique strata/PSU codes
m2['svy_wt'] = m2['wt_itvex'] / m2['year'].nunique()
m2['svy_strata'] = pd.factorize(
    m2['year'].astype(int).astype(str) + '_' + m2['kstrata'].astype(int).astype(str)
)[0] + 1
m2['svy_psu'] = pd.factorize(
    m2['year'].astype(int).astype(str) + '_'
    + m2['kstrata'].astype(int).astype(str) + '_'
    + m2['psu'].astype(str)
)[0] + 1

print(f"Complete-case n = {len(m2)}  |  tooth-loss prevalence = {m2['O_TC_'].mean():.1%}")

# occp is nominal -> dummy-encode (lowest code = reference), replacing the single term
model_vars2 = list(prespec_vars)
# Explicit reference level rather than drop_first, which silently takes the lowest
# OBSERVED code -- that can differ between the two cohort notebooks and change what
# every occp coefficient means. occp 1 = Managers/Professionals, matching §11 and §6.
_nom_d2 = pd.get_dummies(m2['occp'].round().astype(int), prefix='occp', dtype=float)
assert 'occp_1' in _nom_d2.columns, "occp reference level 1 absent from §11.3 sample"
_nom_d2 = _nom_d2.drop(columns=['occp_1'])
m2 = pd.concat([m2, _nom_d2], axis=1)
model_vars2 = [c for c in model_vars2 if c != 'occp'] + list(_nom_d2.columns)

# Age as a RESTRICTED CUBIC SPLINE -- same basis and knots as §11 (see the knot-selection
# cell). This is the CONFIRMATORY model, so the linear-in-age assumption mattered most here:
# age is its strongest predictor and spans ~30 years. 'age' remains as the spline's linear
# term; only the nonlinear columns are added.
m2, model_vars2, _age_nl2 = add_age_spline(m2, model_vars2, AGE_KNOTS)
print(f"  age spline: +{len(_age_nl2)} nonlinear term(s) {_age_nl2}"
      f"  (knots at {', '.join(f'{v:.1f}' for v in AGE_KNOTS)})")

# Drop any zero-variance columns (inestimable in a GLM)
zero_var2 = [c for c in model_vars2 if m2[c].nunique() <= 1 or m2[c].std() < 1e-8]
if zero_var2:
    print(f"Dropping zero-variance columns (inestimable): {zero_var2}")
    model_vars2 = [c for c in model_vars2 if c not in zero_var2]

# Design-based survey logistic regression (survey::svyglm, quasibinomial) ------
import rpy2.robjects as ro
from rpy2.robjects import pandas2ri, numpy2ri
from rpy2.robjects.conversion import localconverter
from rpy2.robjects.packages import importr

survey_pkg = importr('survey')

r_cols2 = ['O_TC_', 'svy_wt', 'svy_strata', 'svy_psu'] + model_vars2
with localconverter(ro.default_converter + numpy2ri.converter + pandas2ri.converter):
    r_df2 = pandas2ri.py2rpy(m2[r_cols2])

ro.r('options(survey.lonely.psu = "adjust")')
svy_design2 = survey_pkg.svydesign(
    ids=ro.Formula('~svy_psu'),
    strata=ro.Formula('~svy_strata'),
    weights=ro.Formula('~svy_wt'),
    data=r_df2
)

formula_str2 = 'O_TC_ ~ ' + ' + '.join(model_vars2)
svyfit2 = survey_pkg.svyglm(
    ro.Formula(formula_str2),
    design=svy_design2,
    family=ro.r('quasibinomial(link="logit")')
)

coef_mat2 = ro.r('summary')(svyfit2).rx2('coefficients')
ci_mat2 = ro.r('confint')(svyfit2, level=0.95)
with localconverter(ro.default_converter + numpy2ri.converter + pandas2ri.converter):
    coef_df2 = pd.DataFrame(
        pandas2ri.rpy2py(coef_mat2),
        index=list(ro.r('rownames')(coef_mat2)),
        columns=['Estimate', 'SE', 't', 'p-value']
    )
    ci_df2 = pd.DataFrame(
        pandas2ri.rpy2py(ci_mat2),
        index=list(ro.r('rownames')(ci_mat2)),
        columns=['CI 2.5%', 'CI 97.5%']
    )

or_table_prespec = pd.DataFrame({
    'aOR':      np.exp(coef_df2['Estimate']),
    'CI 2.5%':  np.exp(ci_df2['CI 2.5%']),
    'CI 97.5%': np.exp(ci_df2['CI 97.5%']),
    'p-value':  coef_df2['p-value']
})

print("\n=== Pre-specified survey-weighted LR (a priori covariates; survey::svyglm) ===")
print(or_table_prespec.round(4).to_string())

print()
wald_nonlinearity(svyfit2, _age_nl2, '§11.3 pre-specified model')
print("The 'age' row above is the spline's LINEAR term, not the whole age effect; read it"
      "\ntogether with the age_rcs* rows and the nonlinearity test.")
```

<h3 style="color:#87CEEB";> 11.3.2 Collinearity Check (VIF) </h3>

Computed on the **pre-specified** model (`m2` / `model_vars2`, §11.3) — the model whose adjusted odds ratios are actually reported, so this is the collinearity that matters for the covariate table. eGFR is derived using age, so we verify it is not collinear with the `age` covariate. VIF must be computed **with an intercept** — omitting it inflates VIF for variables whose mean is far from zero. Rule of thumb: VIF > 5 warrants attention, > 10 is serious.

```python
# Pre-specified model (m2 / model_vars2), not §11's full exploratory frame: this is the
# model whose aORs are reported, so its collinearity is the one that needs checking.
#
# The two NONLINEAR age-spline columns (age_rcs2, age_rcs3) are excluded here: they are
# deterministic basis-function transforms of 'age' by construction, so a high VIF among
# them is guaranteed and uninformative regardless of the rest of the model. 'age' itself
# STAYS IN the check -- the header's eGFR-vs-age collinearity check needs it, since eGFR
# is derived from age via the CKD-EPI equation.
_age_nl_terms = [c for c in model_vars2 if c.startswith('age_rcs')]
vif_vars = [c for c in model_vars2 if c not in _age_nl_terms]

Xc = sm.add_constant(m2[vif_vars].astype(float), has_constant='add')
vif = pd.Series(
    [variance_inflation_factor(Xc.values, i) for i in range(Xc.shape[1])],
    index=Xc.columns
).drop('const')

# For each predictor, its two most-correlated OTHER predictors (Pearson, on the same
# design matrix VIF was computed from). VIF alone flags THAT a term is collinear with the
# rest of the model jointly, not WITH WHAT -- pairwise correlation names actual culprits,
# at the cost of not capturing collinearity with a COMBINATION of others that isn't
# strongly correlated with any single one.
_corr = Xc.drop(columns='const').corr()
_top2 = {}
for col in _corr.columns:
    others = _corr[col].drop(index=col).abs().sort_values(ascending=False)
    _top2[col] = ', '.join(f"{name} ({_corr.loc[col, name]:+.2f})" for name in others.index[:2])

vif_table = pd.DataFrame({'VIF': vif, 'Most correlated with': pd.Series(_top2)}).sort_values('VIF', ascending=False)
print(f"VIF (with intercept) — pre-specified model, {len(vif_vars)} terms "
      f"(age_rcs2/age_rcs3 excluded — collinear with age by construction), n = {len(m2)}:")
print(vif_table.round({'VIF': 2}).to_string())

# C14: guard against the "serious" VIF threshold from the rule-of-thumb in §11.3.2.
# Printed only (not dropped/asserted) -- a flagged predictor doesn't invalidate the
# whole model, but its own aOR/CI is unstable and should be read with that caveat.
_vif_serious  = vif[vif > 10]
_vif_moderate = vif[(vif > 5) & (vif <= 10)]
if not _vif_serious.empty:
    print(f"\nWARNING: VIF > 10 (serious) for: {_vif_serious.round(2).to_dict()}")
    print("  -> aOR/CI for these variables are unstable; interpret with caution.")
if not _vif_moderate.empty:
    print(f"NOTE: VIF > 5 (moderate) for: {_vif_moderate.round(2).to_dict()}")
```

<h3 style="color:#87CEEB";> 11.3.3 Analytic-Sample Composition & Dietary Sensitivity Refit </h3>

Two checks on the complete-case restriction that §11 and §11.3 rely on.

**Why this is needed.** Dietary 24-hour recall coverage is far worse in the COVID-era waves than in the most recent waves (see the per-wave missingness printed below) — so the complete-case filter does **not** remove rows uniformly across waves. The KNHANES pooled weight (annual weight ÷ number of waves) remains correct, since it describes population representation rather than sample composition; the problem is that the *analytic sample* ends up wave-skewed while complete-case inference assumes the missingness is ignorable. §6.1 cannot detect this because it is computed on `df_subset`, **before** the complete-case filter is applied.

1. **Retention by wave** — how many eligible respondents each model actually keeps, and how the wave shares shift.
2. **Included vs excluded (SMD)** — whether the filter shifts the covariate *or outcome* distribution. An `|SMD| > 0.1` on `O_TC_` would mean the exclusion is outcome-related, which is the serious case.
3. **Dietary sensitivity refit** — `N_TDF` and `N_CA` carry most of the wave-dependent missingness. Refitting §11.3 without them recovers the dropped rows; if the remaining aORs are stable, the dietary-driven exclusion is not distorting the confirmatory conclusions.

```python
# ── §11.3.3 Analytic-sample composition & dietary sensitivity ────────────────
# Depends on §11 (`m`) and §11.3 (`m2`, `reg2`, `prespec_vars`, `design_cols`, `survey_pkg`).

# ---- 1. Retention by survey wave -------------------------------------------------
_elig = df_subset.groupby('year').size()
_ret = pd.DataFrame({
    'eligible':        _elig,
    'n §11 (full)':    m.groupby('year').size(),
    'n §11.3 (a priori)': m2.groupby('year').size(),
}).fillna(0).astype(int)
_ret['§11 kept %']   = (_ret['n §11 (full)']       / _ret['eligible'] * 100).round(1)
_ret['§11.3 kept %'] = (_ret['n §11.3 (a priori)'] / _ret['eligible'] * 100).round(1)
_ret['wave share: eligible %'] = (_ret['eligible'] / _ret['eligible'].sum() * 100).round(1)
_ret['wave share: §11.3 %']    = (_ret['n §11.3 (a priori)'] / _ret['n §11.3 (a priori)'].sum() * 100).round(1)

print("── 1. Complete-case retention by survey wave ───────────────────────────────")
print(_ret.to_string())
print(f"\nRetention spread:  §11 {_ret['§11 kept %'].min():.1f}%–{_ret['§11 kept %'].max():.1f}%"
      f"   |   §11.3 {_ret['§11.3 kept %'].min():.1f}%–{_ret['§11.3 kept %'].max():.1f}%")
print("A wide spread means the analytic sample under-represents the low-retention waves"
      "\nrelative to the population the survey weights describe.")

# ---- 2. Included vs excluded rows: standardized mean differences -----------------
# Self-contained helpers (do not depend on §6.1 having been run).
def _smd_c(a, b):
    a, b = a.dropna(), b.dropna()
    if len(a) < 2 or len(b) < 2:
        return np.nan
    pooled = np.sqrt((a.std(ddof=1) ** 2 + b.std(ddof=1) ** 2) / 2)
    return np.nan if pooled == 0 else (a.mean() - b.mean()) / pooled

def _smd_b(a, b, pos):
    a, b = a.dropna(), b.dropna()
    if not len(a) or not len(b):
        return np.nan
    p1, p2 = (a == pos).mean(), (b == pos).mean()
    pooled = np.sqrt((p1 * (1 - p1) + p2 * (1 - p2)) / 2)
    return np.nan if pooled == 0 else (p1 - p2) / pooled

_in_mask = df_subset.index.isin(m2.index)
_incl, _excl = df_subset.loc[_in_mask], df_subset.loc[~_in_mask]

_rows_smd = []
for _c in ['O_TC', 'age', 'eGFR', 'HE_BMI', 'HE_HbA1c', 'HE_sbp', 'N_TDF', 'N_CA']:
    if _c in df_subset.columns:
        _rows_smd.append({'Variable': _c, 'Included': f"{_incl[_c].mean():.2f}",
                          'Excluded': f"{_excl[_c].mean():.2f}",
                          'SMD': _smd_c(_incl[_c], _excl[_c])})
for _c, _pos in [('O_TC_', 1), ('sex', 2), ('diabetes', 1.0), ('sm_presnt', 1),
                 ('town_t', 1), ('BM1_0', 0)]:
    if _c in df_subset.columns:
        _rows_smd.append({'Variable': f"{_c} == {_pos}",
                          'Included': f"{(_incl[_c] == _pos).mean() * 100:.1f}%",
                          'Excluded': f"{(_excl[_c] == _pos).mean() * 100:.1f}%",
                          'SMD': _smd_b(_incl[_c], _excl[_c], _pos)})
_smd_df = pd.DataFrame(_rows_smd)
_smd_df['Flag'] = np.where(_smd_df['SMD'].abs() > 0.1, '*', '')

print("\n── 2. §11.3 analytic sample vs EXCLUDED rows (|SMD| > 0.1 flagged *) ───────")
print(f"included n = {len(_incl)}   excluded n = {len(_excl)}"
      f"   ({len(_excl) / len(df_subset) * 100:.1f}% of the cohort dropped)")
print(_smd_df.round(3).to_string(index=False))
_out_smd = _smd_df.loc[_smd_df['Variable'] == 'O_TC_ == 1', 'SMD']
if len(_out_smd) and abs(_out_smd.iloc[0]) > 0.1:
    print(f"\nWARNING: outcome SMD = {_out_smd.iloc[0]:+.3f} (|SMD| > 0.1) -- the complete-case"
          "\nfilter is OUTCOME-RELATED, so the aORs above are subject to selection bias.")
else:
    print("\nOutcome SMD within +/-0.1: no evidence the exclusion is outcome-related.")

# ---- 3. §11.3 refit WITHOUT the dietary block ------------------------------------
prespec_nodiet = [v for v in prespec_vars if v not in ('N_TDF', 'N_CA')]
m2b = (reg2[['O_TC_'] + design_cols + prespec_nodiet]
       .replace([np.inf, -np.inf], np.nan).dropna())
m2b['svy_wt'] = m2b['wt_itvex'] / m2b['year'].nunique()
m2b['svy_strata'] = pd.factorize(
    m2b['year'].astype(int).astype(str) + '_' + m2b['kstrata'].astype(int).astype(str)
)[0] + 1
m2b['svy_psu'] = pd.factorize(
    m2b['year'].astype(int).astype(str) + '_'
    + m2b['kstrata'].astype(int).astype(str) + '_'
    + m2b['psu'].astype(str)
)[0] + 1

model_vars2b = list(prespec_nodiet)
_nom_d2b = pd.get_dummies(m2b['occp'].round().astype(int), prefix='occp', dtype=float)
assert 'occp_1' in _nom_d2b.columns, "occp reference level 1 absent"
_nom_d2b = _nom_d2b.drop(columns=['occp_1'])
m2b = pd.concat([m2b, _nom_d2b], axis=1)
model_vars2b = [c for c in model_vars2b if c != 'occp'] + list(_nom_d2b.columns)
# Same age spline as §11.3, so the with/without-diet comparison differs ONLY in the dietary
# block. Leaving this refit linear-in-age would confound the diet effect with the age term.
m2b, model_vars2b, _ = add_age_spline(m2b, model_vars2b, AGE_KNOTS)
model_vars2b = [c for c in model_vars2b if m2b[c].nunique() > 1 and m2b[c].std() >= 1e-8]

print(f"\n── 3. §11.3 refit without the dietary block (N_TDF, N_CA) ──────────────────")
print(f"n = {len(m2b)} (vs {len(m2)} with diet; {len(m2b) - len(m2):+d} rows)"
      f"  |  prevalence = {m2b['O_TC_'].mean():.1%} (vs {m2['O_TC_'].mean():.1%})")

r_cols2b = ['O_TC_', 'svy_wt', 'svy_strata', 'svy_psu'] + model_vars2b
with localconverter(ro.default_converter + numpy2ri.converter + pandas2ri.converter):
    r_df2b = pandas2ri.py2rpy(m2b[r_cols2b])

ro.r('options(survey.lonely.psu = "adjust")')
svy_design2b = survey_pkg.svydesign(ids=ro.Formula('~svy_psu'),
                                    strata=ro.Formula('~svy_strata'),
                                    weights=ro.Formula('~svy_wt'), data=r_df2b)
svyfit2b = survey_pkg.svyglm(
    ro.Formula('O_TC_ ~ ' + ' + '.join(model_vars2b)),
    design=svy_design2b, family=ro.r('quasibinomial(link="logit")'))

_cm2b = ro.r('summary')(svyfit2b).rx2('coefficients')
_ci2b = ro.r('confint')(svyfit2b, level=0.95)
with localconverter(ro.default_converter + numpy2ri.converter + pandas2ri.converter):
    _cdf2b = pd.DataFrame(pandas2ri.rpy2py(_cm2b), index=list(ro.r('rownames')(_cm2b)),
                          columns=['Estimate', 'SE', 't', 'p-value'])
    _cidf2b = pd.DataFrame(pandas2ri.rpy2py(_ci2b), index=list(ro.r('rownames')(_ci2b)),
                           columns=['CI 2.5%', 'CI 97.5%'])

or_table_nodiet = pd.DataFrame({
    'aOR': np.exp(_cdf2b['Estimate']),
    'CI 2.5%': np.exp(_cidf2b['CI 2.5%']),
    'CI 97.5%': np.exp(_cidf2b['CI 97.5%']),
    'p-value': _cdf2b['p-value'],
})

_shared = [v for v in or_table_prespec.index if v in or_table_nodiet.index and v != '(Intercept)']
_cmp = pd.DataFrame({
    'aOR with diet': or_table_prespec.loc[_shared, 'aOR'],
    'aOR no diet':   or_table_nodiet.loc[_shared, 'aOR'],
    'p with diet':   or_table_prespec.loc[_shared, 'p-value'],
    'p no diet':     or_table_nodiet.loc[_shared, 'p-value'],
})
_cmp['aOR ratio'] = _cmp['aOR no diet'] / _cmp['aOR with diet']
# Direction flip only counts when at least one of the two estimates is significant.
# A null aOR wobbling from 1.0006 to 0.9965 crosses 1.0 but means nothing, and flagging
# it would bury the flips that actually matter.
_cmp['direction flip'] = np.where(
    ((_cmp['aOR with diet'] - 1) * (_cmp['aOR no diet'] - 1) < 0)
    & ((_cmp['p with diet'] < 0.05) | (_cmp['p no diet'] < 0.05)), 'YES', '')
_cmp['sig change'] = np.where(
    (_cmp['p with diet'] < 0.05) != (_cmp['p no diet'] < 0.05), 'YES', '')

print(_cmp.round(4).to_string())
_unstable = _cmp[(_cmp['direction flip'] == 'YES') | (_cmp['sig change'] == 'YES')]
if len(_unstable):
    print(f"\n{len(_unstable)} coefficient(s) change direction or significance when the dietary"
          "\nblock is removed -- report these as sensitive to the dietary-driven exclusion:")
    print("  " + ", ".join(_unstable.index))
else:
    print("\nNo coefficient changes direction or crosses p=0.05 when the dietary block is"
          "\nremoved -- the confirmatory conclusions are stable to the wave-skewed exclusion.")
```
